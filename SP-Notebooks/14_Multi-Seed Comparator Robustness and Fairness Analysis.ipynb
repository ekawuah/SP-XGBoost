```python
# ======================================================================
# NOTEBOOK 14 — MULTI-SEED COMPARATOR ROBUSTNESS
#               AND FAIRNESS ANALYSIS
#
# CELL 4 — COMPARATOR CONFIGURATION
#          AND TUNING-FAIRNESS AUDIT
#
# DIAGNOSTIC ONLY — NO MODEL TRAINING
# ======================================================================

print("=" * 70)
print("NOTEBOOK 14 — COMPARATOR CONFIGURATION")
print("AND TUNING-FAIRNESS AUDIT")
print("=" * 70)

# ----------------------------------------------------------------------
# REQUIRED IMPORTS
# ----------------------------------------------------------------------

import json
import pickle
import re

from pathlib import Path

import numpy as np
import pandas as pd

print()
print("✓ Diagnostic libraries imported")

# ======================================================================
# 1. DEFINE COMPARATOR IDENTIFIERS
# ======================================================================

COMPARATOR_NAMES = [
    "Logistic Regression",
    "Random Forest",
    "Support Vector Machine"
]

COMPARATOR_KEYWORDS = {
    "Logistic Regression": [
        "logistic",
        "logistic_regression",
        "logisticregression",
        "lr"
    ],

    "Random Forest": [
        "random_forest",
        "randomforest",
        "rf"
    ],

    "Support Vector Machine": [
        "support_vector",
        "supportvectormachine",
        "svm",
        "svc"
    ]
}

print()
print("=" * 70)
print("COMPARATOR MODELS UNDER AUDIT")
print("=" * 70)

for model_name in COMPARATOR_NAMES:
    print(f"✓ {model_name}")

# ======================================================================
# 2. SEARCH RESULTS CSV FILES
# ======================================================================

print()
print("=" * 70)
print("SEARCHING RESULT CSV ARTIFACTS")
print("=" * 70)

all_result_csvs = sorted(
    RESULTS_DIR.rglob("*.csv")
)

print(
    f"CSV artifacts found: "
    f"{len(all_result_csvs)}"
)

comparator_csv_candidates = []

for csv_path in all_result_csvs:

    filename_lower = (
        csv_path.name
        .lower()
        .replace("-", "_")
        .replace(" ", "_")
    )

    matched_models = []

    for model_name, keywords in (
        COMPARATOR_KEYWORDS.items()
    ):

        if any(
            keyword in filename_lower
            for keyword in keywords
        ):

            matched_models.append(
                model_name
            )

    # Also inspect Model column if available
    try:

        df_check = pd.read_csv(
            csv_path
        )

        if "Model" in df_check.columns:

            model_values = (
                df_check["Model"]
                .astype(str)
                .tolist()
            )

            for model_name in COMPARATOR_NAMES:

                if any(
                    model_name.lower()
                    in value.lower()
                    for value in model_values
                ):

                    if model_name not in matched_models:
                        matched_models.append(
                            model_name
                        )

            # SVM abbreviation
            if any(
                value.strip().upper()
                == "SVM"
                for value in model_values
            ):

                if (
                    "Support Vector Machine"
                    not in matched_models
                ):

                    matched_models.append(
                        "Support Vector Machine"
                    )

        columns_found = list(
            df_check.columns
        )

    except Exception as e:

        columns_found = [
            f"READ ERROR: {e}"
        ]

    if len(matched_models) > 0:

        comparator_csv_candidates.append({
            "Path": str(csv_path),
            "Models": matched_models,
            "Columns": columns_found
        })

# ----------------------------------------------------------------------
# DISPLAY CSV CANDIDATES
# ----------------------------------------------------------------------

if len(comparator_csv_candidates) == 0:

    print(
        "⚠ No comparator-specific CSV artifacts found"
    )

else:

    for i, artifact in enumerate(
        comparator_csv_candidates,
        start=1
    ):

        print()
        print(
            f"[CSV {i}] "
            f"{artifact['Path']}"
        )

        print(
            f"Models: "
            f"{artifact['Models']}"
        )

        print(
            f"Columns: "
            f"{artifact['Columns']}"
        )

# ======================================================================
# 3. SEARCH SAVED MODEL FILES
# ======================================================================

print()
print("=" * 70)
print("SEARCHING SAVED MODEL ARTIFACTS")
print("=" * 70)

MODEL_EXTENSIONS = [
    "*.pkl",
    "*.pickle",
    "*.joblib"
]

model_files = []

for extension in MODEL_EXTENSIONS:

    model_files.extend(
        MODEL_DIR.rglob(
            extension
        )
    )

model_files = sorted(
    set(model_files)
)

print(
    f"Saved estimator files found: "
    f"{len(model_files)}"
)

comparator_model_candidates = []

for model_path in model_files:

    filename_lower = (
        model_path.name
        .lower()
        .replace("-", "_")
        .replace(" ", "_")
    )

    matched_models = []

    for model_name, keywords in (
        COMPARATOR_KEYWORDS.items()
    ):

        if any(
            keyword in filename_lower
            for keyword in keywords
        ):

            matched_models.append(
                model_name
            )

    if len(matched_models) > 0:

        comparator_model_candidates.append({
            "Path": model_path,
            "Matched_Models": matched_models
        })

# ----------------------------------------------------------------------
# DISPLAY SAVED MODEL CANDIDATES
# ----------------------------------------------------------------------

if len(comparator_model_candidates) == 0:

    print(
        "⚠ No comparator model files identified "
        "from filenames"
    )

else:

    for i, candidate in enumerate(
        comparator_model_candidates,
        start=1
    ):

        print()
        print(
            f"[MODEL {i}] "
            f"{candidate['Path']}"
        )

        print(
            f"Possible model: "
            f"{candidate['Matched_Models']}"
        )

# ======================================================================
# 4. ATTEMPT TO INSPECT SAVED ESTIMATORS
# ======================================================================

print()
print("=" * 70)
print("SAVED ESTIMATOR PARAMETER INSPECTION")
print("=" * 70)

saved_model_parameter_records = []

for candidate in comparator_model_candidates:

    model_path = candidate["Path"]

    try:

        # --------------------------------------------------------------
        # Try joblib first if available
        # --------------------------------------------------------------

        try:

            import joblib

            loaded_object = joblib.load(
                model_path
            )

        except Exception:

            with open(
                model_path,
                "rb"
            ) as file:

                loaded_object = pickle.load(
                    file
                )

        # --------------------------------------------------------------
        # Determine estimator structure
        # --------------------------------------------------------------

        object_type = type(
            loaded_object
        ).__name__

        print()
        print("-" * 70)
        print(
            f"File: {model_path}"
        )

        print(
            f"Object type: "
            f"{object_type}"
        )

        # --------------------------------------------------------------
        # Inspect sklearn pipeline
        # --------------------------------------------------------------

        if hasattr(
            loaded_object,
            "named_steps"
        ):

            print(
                "Pipeline steps:"
            )

            for step_name, step_object in (
                loaded_object
                .named_steps
                .items()
            ):

                print(
                    f"  {step_name}: "
                    f"{type(step_object).__name__}"
                )

            estimator = list(
                loaded_object
                .named_steps
                .values()
            )[-1]

        else:

            estimator = loaded_object

        estimator_type = type(
            estimator
        ).__name__

        print(
            f"Final estimator: "
            f"{estimator_type}"
        )

        # --------------------------------------------------------------
        # Extract parameters
        # --------------------------------------------------------------

        if hasattr(
            estimator,
            "get_params"
        ):

            params = estimator.get_params(
                deep=False
            )

            saved_model_parameter_records.append({
                "File":
                    str(model_path),

                "Estimator":
                    estimator_type,

                "Parameters":
                    params
            })

            print(
                "Estimator parameters:"
            )

            for key in sorted(
                params.keys()
            ):

                print(
                    f"  {key}: "
                    f"{params[key]}"
                )

        else:

            print(
                "⚠ Estimator does not expose "
                "get_params()"
            )

    except Exception as e:

        print()
        print(
            f"⚠ Could not inspect "
            f"{model_path}"
        )

        print(
            f"  Error: {e}"
        )

# ======================================================================
# 5. SEARCH PROJECT NOTEBOOKS FOR COMPARATOR CONFIGURATION
# ======================================================================

print()
print("=" * 70)
print("SEARCHING NOTEBOOKS FOR COMPARATOR CONFIGURATION")
print("=" * 70)

notebook_files = sorted(
    PROJECT_DIR.rglob(
        "*.ipynb"
    )
)

print(
    f"Notebook files found: "
    f"{len(notebook_files)}"
)

# ----------------------------------------------------------------------
# Prefer files likely to correspond to Notebook 9
# ----------------------------------------------------------------------

notebook9_candidates = [
    path
    for path in notebook_files
    if (
        "9" in path.stem.lower()
        or
        "comparator" in path.stem.lower()
        or
        "benchmark" in path.stem.lower()
    )
]

print()
print(
    f"Potential Notebook 9/comparator notebooks: "
    f"{len(notebook9_candidates)}"
)

for path in notebook9_candidates:

    print(
        f"✓ {path}"
    )

# ----------------------------------------------------------------------
# Search notebook source code
# ----------------------------------------------------------------------

SEARCH_TERMS = [
    "LogisticRegression",
    "RandomForestClassifier",
    "SVC",
    "GridSearchCV",
    "RandomizedSearchCV",
    "cross_val_score",
    "StratifiedKFold",
    "StandardScaler",
    "C=",
    "n_estimators",
    "max_depth"
]

notebook_hits = []

for notebook_path in notebook9_candidates:

    try:

        with open(
            notebook_path,
            "r",
            encoding="utf-8"
        ) as file:

            notebook_json = json.load(
                file
            )

        source_lines = []

        for cell_number, cell in enumerate(
            notebook_json.get(
                "cells",
                []
            ),
            start=1
        ):

            source = "".join(
                cell.get(
                    "source",
                    []
                )
            )

            if cell.get(
                "cell_type"
            ) != "code":

                continue

            matched_terms = [
                term
                for term in SEARCH_TERMS
                if term.lower()
                in source.lower()
            ]

            if len(matched_terms) > 0:

                notebook_hits.append({
                    "Notebook":
                        str(notebook_path),

                    "Cell":
                        cell_number,

                    "Terms":
                        matched_terms,

                    "Source":
                        source
                })

    except Exception:
        pass

# ----------------------------------------------------------------------
# DISPLAY RELEVANT NOTEBOOK CODE
# ----------------------------------------------------------------------

if len(notebook_hits) == 0:

    print()
    print(
        "⚠ No comparator-configuration code "
        "was recovered from candidate notebooks"
    )

else:

    print()
    print("=" * 70)
    print("COMPARATOR CONFIGURATION CODE FOUND")
    print("=" * 70)

    for hit in notebook_hits:

        print()
        print("-" * 70)

        print(
            f"Notebook: "
            f"{hit['Notebook']}"
        )

        print(
            f"Cell: "
            f"{hit['Cell']}"
        )

        print(
            f"Matched terms: "
            f"{hit['Terms']}"
        )

        print()
        print(
            hit["Source"]
        )

# ======================================================================
# 6. LOCKED NOTEBOOK 13 COMPARATOR PERFORMANCE
# ======================================================================

print()
print("=" * 70)
print("LOCKED NOTEBOOK 13 COMPARATOR PERFORMANCE")
print("=" * 70)

final_six_model_df = pd.read_csv(
    FINAL_SIX_MODEL_PATH
)

required_comparator_models = [
    "Logistic Regression",
    "Random Forest",
    "Support Vector Machine"
]

for model_name in required_comparator_models:

    model_rows = (
        final_six_model_df.loc[
            final_six_model_df[
                "Model"
            ] == model_name
        ]
    )

    assert len(
        model_rows
    ) == 1

print(
    final_six_model_df.loc[
        final_six_model_df["Model"]
        .isin(required_comparator_models)
    ].to_string(
        index=False
    )
)

print()
print(
    "✓ Existing comparator performance "
    "recovered for provenance only"
)

# ======================================================================
# 7. TUNING-EVIDENCE AUDIT
# ======================================================================

print()
print("=" * 70)
print("TUNING-EVIDENCE AUDIT")
print("=" * 70)

# ----------------------------------------------------------------------
# Detect common tuning terms in recovered notebook code
# ----------------------------------------------------------------------

all_recovered_code = "\n".join(
    hit["Source"]
    for hit in notebook_hits
)

tuning_indicators = {
    "GridSearchCV":
        "gridsearchcv"
        in all_recovered_code.lower(),

    "RandomizedSearchCV":
        "randomizedsearchcv"
        in all_recovered_code.lower(),

    "Optuna":
        "optuna"
        in all_recovered_code.lower(),

    "Manual parameter grid":
        "param_grid"
        in all_recovered_code.lower(),

    "5-fold CV":
        (
            "n_splits=5"
            in all_recovered_code.replace(
                " ",
                ""
            ).lower()
            or
            "cv=5"
            in all_recovered_code.replace(
                " ",
                ""
            ).lower()
        )
}

for evidence, present in (
    tuning_indicators.items()
):

    print(
        f"{evidence:<25}: "
        f"{present}"
    )

# ======================================================================
# 8. FAIRNESS STATUS
# ======================================================================

print()
print("=" * 70)
print("PRELIMINARY FAIRNESS STATUS")
print("=" * 70)

print(
    "The following must be established "
    "before any comparator retraining:"
)

print()
print(
    "1. Exact LR configuration"
)

print(
    "2. Exact RF configuration"
)

print(
    "3. Exact SVM configuration"
)

print(
    "4. Scaling procedure for LR/SVM"
)

print(
    "5. Hyperparameter search method per model"
)

print(
    "6. Number of candidate configurations/trials per model"
)

print(
    "7. Cross-validation folds used during tuning"
)

print(
    "8. Primary tuning metric"
)

print(
    "9. Whether all models had equivalent access "
    "to the 49-feature training data"
)

print(
    "10. Whether tuning-budget parity was actually satisfied"
)

# ----------------------------------------------------------------------
# IMPORTANT: DO NOT ASSERT FAIRNESS YET
# ----------------------------------------------------------------------

FAIRNESS_CONFIRMED = False

print()
print(
    f"FAIRNESS_CONFIRMED = "
    f"{FAIRNESS_CONFIRMED}"
)

print()
print(
    "⚠ Comparator fairness is intentionally "
    "NOT declared in Cell 4."
)

print(
    "⚠ Review the recovered configurations "
    "before selecting the Notebook 14 protocol."
)

# ======================================================================
# FINAL VALIDATION
# ======================================================================

print()
print("=" * 70)
print("CELL 4 VALIDATION")
print("=" * 70)

print(
    "✓ Result CSV artifacts inspected"
)

print(
    "✓ Saved comparator model artifacts searched"
)

print(
    "✓ Saved estimator parameters inspected where available"
)

print(
    "✓ Comparator/notebook source code searched"
)

print(
    "✓ Existing comparator performance recovered"
)

print(
    "✓ Tuning evidence audited"
)

print(
    "✓ Fairness has NOT been assumed"
)

print(
    "✓ No model trained"
)

print(
    "✓ No hyperparameter search performed"
)

print(
    "✓ No prediction generated"
)

print(
    "✓ Independent test set remains isolated"
)

print(
    "✓ No previous artifact modified"
)

print(
    "✓ Notebooks 11, 12, and 13 remain locked"
)

print()
print("=" * 70)
print("✓ CELL 4 — FAIRNESS AUDIT COMPLETED")
print("=" * 70)
```

```python
# ======================================================================
# NOTEBOOK 14 — MULTI-SEED COMPARATOR ROBUSTNESS
#               AND FAIRNESS ANALYSIS
## CELL 6 — THREE-SEED LOGISTIC REGRESSION ROBUSTNESS
# ======================================================================

print("=" * 70)
print("NOTEBOOK 14 — THREE-SEED LOGISTIC REGRESSION ROBUSTNESS")
print("=" * 70)

# ----------------------------------------------------------------------
# REQUIRED IMPORTS
# ----------------------------------------------------------------------

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import StratifiedKFold, GridSearchCV

from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    f1_score,
    precision_score,
    recall_score,
    accuracy_score,
    confusion_matrix
)

print()
print("✓ Logistic Regression evaluation libraries imported")

# ======================================================================
# 1. LOCKED LOGISTIC REGRESSION SEARCH SPACE
# ======================================================================

LOGISTIC_GRID = {
    "classifier__C": [
        0.01,
        0.1,
        1.0,
        10.0,
        100.0
    ],
    "classifier__solver": [
        "lbfgs"
    ]
}

assert len(
    LOGISTIC_GRID["classifier__C"]
) == 5

print()
print("=" * 70)
print("LOCKED LOGISTIC REGRESSION SEARCH SPACE")
print("=" * 70)

print("C values:")
print(LOGISTIC_GRID["classifier__C"])

print()
print("Solver:")
print(LOGISTIC_GRID["classifier__solver"])

print()
print("✓ 5 candidate configurations confirmed")
print("✓ StandardScaler will be fitted inside each CV pipeline")
print("✓ ROC-AUC used for hyperparameter selection")

# ======================================================================
# 2. INPUT VALIDATION
# ======================================================================

assert X_train_49.shape == (468, 49)
assert X_test_49.shape == (117, 49)

assert y_train_49.shape == (468,)
assert y_test_49.shape == (117,)

assert RANDOM_SEEDS == [42, 123, 7]

print()
print("=" * 70)
print("INPUT VALIDATION")
print("=" * 70)

print("✓ Training matrix: 468 × 49")
print("✓ Independent test matrix: 117 × 49")
print("✓ Three random seeds confirmed")
print("✓ Test set remains isolated from tuning")

# ======================================================================
# 3. RESULT CONTAINERS
# ======================================================================

lr_seed_results = []
lr_best_params = []
lr_seed_models = {}
lr_seed_predictions = {}
lr_seed_probabilities = {}

# ======================================================================
# 4. THREE-SEED TUNING AND FINAL TRAINING
# ======================================================================

for seed in RANDOM_SEEDS:

    print()
    print("=" * 70)
    print(f"LOGISTIC REGRESSION — RANDOM SEED {seed}")
    print("=" * 70)

    # --------------------------------------------------------------
    # Seed-specific CV splitter
    # --------------------------------------------------------------

    cv = StratifiedKFold(
        n_splits=5,
        shuffle=True,
        random_state=seed
    )

    # --------------------------------------------------------------
    # Pipeline
    # --------------------------------------------------------------

    logistic_pipeline = Pipeline([
        (
            "scaler",
            StandardScaler()
        ),
        (
            "classifier",
            LogisticRegression(
                max_iter=5000,
                random_state=seed
            )
        )
    ])

    # --------------------------------------------------------------
    # Grid search on training data only
    # --------------------------------------------------------------

    logistic_search = GridSearchCV(
        estimator=logistic_pipeline,
        param_grid=LOGISTIC_GRID,
        scoring="roc_auc",
        cv=cv,
        n_jobs=-1,
        refit=True,
        return_train_score=False
    )

    logistic_search.fit(
        X_train_49,
        y_train_49
    )

    best_model = (
        logistic_search.best_estimator_
    )

    lr_seed_models[seed] = (
        best_model
    )

    # --------------------------------------------------------------
    # Save best tuning result
    # --------------------------------------------------------------

    lr_best_params.append({
        "Seed": seed,
        "Best_C": (
            logistic_search
            .best_params_[
                "classifier__C"
            ]
        ),
        "Best_Solver": (
            logistic_search
            .best_params_[
                "classifier__solver"
            ]
        ),
        "Best_CV_ROC_AUC": (
            logistic_search.best_score_
        )
    })

    print(
        f"Best parameters: "
        f"{logistic_search.best_params_}"
    )

    print(
        f"Best 5-fold CV ROC-AUC: "
        f"{logistic_search.best_score_:.4f}"
    )

    # --------------------------------------------------------------
    # Independent-test prediction
    # --------------------------------------------------------------

    y_probability = (
        best_model.predict_proba(
            X_test_49
        )[:, 1]
    )

    y_prediction = (
        best_model.predict(
            X_test_49
        )
    )

    lr_seed_probabilities[
        seed
    ] = y_probability

    lr_seed_predictions[
        seed
    ] = y_prediction

    # --------------------------------------------------------------
    # Independent-test metrics
    # --------------------------------------------------------------

    roc_auc = roc_auc_score(
        y_test_49,
        y_probability
    )

    pr_auc = average_precision_score(
        y_test_49,
        y_probability
    )

    f1 = f1_score(
        y_test_49,
        y_prediction
    )

    precision = precision_score(
        y_test_49,
        y_prediction,
        zero_division=0
    )

    recall = recall_score(
        y_test_49,
        y_prediction,
        zero_division=0
    )

    accuracy = accuracy_score(
        y_test_49,
        y_prediction
    )

    tn, fp, fn, tp = (
        confusion_matrix(
            y_test_49,
            y_prediction
        )
        .ravel()
    )

    specificity = (
        tn /
        (tn + fp)
    )

    # --------------------------------------------------------------
    # Store result
    # --------------------------------------------------------------

    lr_seed_results.append({
        "Seed": seed,
        "ROC_AUC": roc_auc,
        "PR_AUC": pr_auc,
        "F1": f1,
        "Precision": precision,
        "Recall": recall,
        "Specificity": specificity,
        "Accuracy": accuracy,
        "TN": int(tn),
        "FP": int(fp),
        "FN": int(fn),
        "TP": int(tp)
    })

    print()
    print(
        f"Independent-test ROC-AUC: "
        f"{roc_auc:.4f}"
    )

    print(
        f"Independent-test PR-AUC:  "
        f"{pr_auc:.4f}"
    )

    print(
        f"Independent-test F1:      "
        f"{f1:.4f}"
    )

# ======================================================================
# 5. CREATE RESULT TABLES
# ======================================================================

lr_seed_results_df = pd.DataFrame(
    lr_seed_results
)

lr_best_params_df = pd.DataFrame(
    lr_best_params
)

# ----------------------------------------------------------------------
# Validation
# ----------------------------------------------------------------------

assert len(
    lr_seed_results_df
) == 3

assert len(
    lr_best_params_df
) == 3

assert set(
    lr_seed_results_df["Seed"]
) == {42, 123, 7}

assert set(
    lr_best_params_df["Seed"]
) == {42, 123, 7}

# ----------------------------------------------------------------------
# All metrics finite
# ----------------------------------------------------------------------

LR_METRICS = [
    "ROC_AUC",
    "PR_AUC",
    "F1",
    "Precision",
    "Recall",
    "Specificity",
    "Accuracy"
]

assert np.isfinite(
    lr_seed_results_df[
        LR_METRICS
    ].to_numpy()
).all()

# ======================================================================
# 6. THREE-SEED ROBUSTNESS SUMMARY
# ======================================================================

lr_robustness_rows = []

for metric in LR_METRICS:

    values = (
        lr_seed_results_df[
            metric
        ]
        .astype(float)
    )

    lr_robustness_rows.append({
        "Metric": metric,
        "Mean": values.mean(),
        "SD": values.std(ddof=1),
        "Minimum": values.min(),
        "Maximum": values.max(),
        "Range": (
            values.max()
            -
            values.min()
        ),
        "CV_Percent": (
            values.std(ddof=1)
            /
            values.mean()
        ) * 100
    })

lr_robustness_summary_df = pd.DataFrame(
    lr_robustness_rows
)

assert len(
    lr_robustness_summary_df
) == 7

# ======================================================================
# 7. DISPLAY RESULTS
# ======================================================================

print()
print("=" * 70)
print("THREE-SEED LOGISTIC REGRESSION TEST RESULTS")
print("=" * 70)

print(
    lr_seed_results_df.to_string(
        index=False,
        formatters={
            metric: "{:.4f}".format
            for metric in LR_METRICS
        }
    )
)

print()
print("=" * 70)
print("THREE-SEED LOGISTIC REGRESSION MEAN ± SD")
print("=" * 70)

for _, row in (
    lr_robustness_summary_df.iterrows()
):

    print(
        f"{row['Metric']:<15}: "
        f"{row['Mean']:.4f} ± "
        f"{row['SD']:.4f}"
    )

# ======================================================================
# 8. SAVE NOTEBOOK 14 ARTIFACTS
# ======================================================================

NOTEBOOK14_DIR = (
    RESULTS_DIR /
    "notebook14_fairness"
)

NOTEBOOK14_DIR.mkdir(
    parents=True,
    exist_ok=True
)

LR_RESULTS_PATH = (
    NOTEBOOK14_DIR /
    "logistic_regression_three_seed_test_results.csv"
)

LR_SUMMARY_PATH = (
    NOTEBOOK14_DIR /
    "logistic_regression_three_seed_robustness_summary.csv"
)

LR_PARAMS_PATH = (
    NOTEBOOK14_DIR /
    "logistic_regression_three_seed_best_params.csv"
)

lr_seed_results_df.to_csv(
    LR_RESULTS_PATH,
    index=False
)

lr_robustness_summary_df.to_csv(
    LR_SUMMARY_PATH,
    index=False
)

lr_best_params_df.to_csv(
    LR_PARAMS_PATH,
    index=False
)

assert LR_RESULTS_PATH.exists()
assert LR_SUMMARY_PATH.exists()
assert LR_PARAMS_PATH.exists()

# ======================================================================
# 9. FINAL VALIDATION
# ======================================================================

print()
print("=" * 70)
print("CELL 6 VALIDATION")
print("=" * 70)

print("✓ Logistic Regression evaluated across three seeds")
print("✓ Seed-specific 5-fold tuning used")
print("✓ Original 5-candidate LR grid preserved")
print("✓ Standardization fitted inside CV pipeline")
print("✓ ROC-AUC used for tuning")
print("✓ Independent test set not used during tuning")
print("✓ Seven independent-test metrics calculated")
print("✓ Confusion matrices calculated")
print("✓ Three-seed mean ± SD calculated")
print("✓ Best parameters saved per seed")
print("✓ Logistic Regression robustness artifacts saved")
print(f"✓ {LR_RESULTS_PATH}")
print(f"✓ {LR_SUMMARY_PATH}")
print(f"✓ {LR_PARAMS_PATH}")
print("✓ No feature selection rerun")
print("✓ No previous notebook artifact modified")
print("✓ Notebooks 11–13 remain locked")

print()
print("=" * 70)
print("✓ CELL 6 — COMPLETED")
print("=" * 70)
```

```python
# ======================================================================
# NOTEBOOK 14 — MULTI-SEED COMPARATOR ROBUSTNESS
#               AND FAIRNESS ANALYSIS
#
# CELL 8 — THREE-SEED SUPPORT VECTOR MACHINE ROBUSTNESS
#          MATCHED 49-FEATURE ANALYSIS
# ======================================================================

print("=" * 70)
print("NOTEBOOK 14 — THREE-SEED SUPPORT VECTOR MACHINE ROBUSTNESS")
print("=" * 70)

# ----------------------------------------------------------------------
# REQUIRED IMPORTS
# ----------------------------------------------------------------------

from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.model_selection import StratifiedKFold, GridSearchCV

from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    f1_score,
    precision_score,
    recall_score,
    accuracy_score,
    confusion_matrix
)

print()
print("✓ Support Vector Machine evaluation libraries imported")

# ======================================================================
# 1. ORIGINAL NOTEBOOK 9 SVM SEARCH SPACE
# ======================================================================

SVM_GRID = {
    "classifier__C": [
        0.1,
        1.0,
        10.0,
        100.0
    ],

    "classifier__kernel": [
        "linear",
        "rbf"
    ]
}

svm_candidate_count = (
    len(SVM_GRID["classifier__C"])
    *
    len(SVM_GRID["classifier__kernel"])
)

assert svm_candidate_count == 8

print()
print("=" * 70)
print("LOCKED SUPPORT VECTOR MACHINE SEARCH SPACE")
print("=" * 70)

print(
    "C values:",
    SVM_GRID["classifier__C"]
)

print(
    "Kernels:",
    SVM_GRID["classifier__kernel"]
)

print()
print(
    f"Total candidate configurations: "
    f"{svm_candidate_count}"
)

print()
print("✓ Original Notebook 9 SVM search space preserved")
print("✓ StandardScaler will be fitted inside each CV pipeline")
print("✓ ROC-AUC used for hyperparameter selection")

# ======================================================================
# 2. MATCHED INPUT VALIDATION
# ======================================================================

assert X_train_49.shape == (468, 49)
assert X_test_49.shape == (117, 49)

assert y_train_49.shape == (468,)
assert y_test_49.shape == (117,)

assert RANDOM_SEEDS == [42, 123, 7]

assert list(
    X_train_49.columns
) == list(
    X_test_49.columns
)

assert np.isfinite(
    X_train_49.to_numpy(
        dtype=float
    )
).all()

assert np.isfinite(
    X_test_49.to_numpy(
        dtype=float
    )
).all()

print()
print("=" * 70)
print("MATCHED 49-FEATURE INPUT VALIDATION")
print("=" * 70)

print("✓ Training matrix: 468 × 49")
print("✓ Independent test matrix: 117 × 49")
print("✓ Same 49-feature space used as Notebook 14 LR and RF")
print("✓ Three random seeds confirmed")
print("✓ All predictor values finite")
print("✓ Independent test set isolated from tuning")

# ======================================================================
# 3. RESULT CONTAINERS
# ======================================================================

svm_seed_results = []

svm_best_params = []

svm_seed_models = {}

svm_seed_predictions = {}

svm_seed_probabilities = {}

# ======================================================================
# 4. THREE-SEED SVM TUNING AND TEST EVALUATION
# ======================================================================

for seed in RANDOM_SEEDS:

    print()
    print("=" * 70)
    print(
        f"SUPPORT VECTOR MACHINE — RANDOM SEED {seed}"
    )
    print("=" * 70)

    # --------------------------------------------------------------
    # Seed-specific stratified 5-fold CV
    # --------------------------------------------------------------

    cv = StratifiedKFold(
        n_splits=5,
        shuffle=True,
        random_state=seed
    )

    # --------------------------------------------------------------
    # Scaling is fitted within each CV fold
    # --------------------------------------------------------------

    svm_pipeline = Pipeline(
        steps=[
            (
                "scaler",
                StandardScaler()
            ),
            (
                "classifier",
                SVC(
                    probability=True,
                    random_state=seed
                )
            )
        ]
    )

    # --------------------------------------------------------------
    # Grid search — TRAINING DATA ONLY
    # --------------------------------------------------------------

    svm_search = GridSearchCV(
        estimator=svm_pipeline,
        param_grid=SVM_GRID,
        scoring="roc_auc",
        cv=cv,
        n_jobs=-1,
        refit=True,
        return_train_score=False
    )

    svm_search.fit(
        X_train_49,
        y_train_49
    )

    best_model = (
        svm_search.best_estimator_
    )

    svm_seed_models[
        seed
    ] = best_model

    # --------------------------------------------------------------
    # Store seed-specific best parameters
    # --------------------------------------------------------------

    svm_best_params.append({

        "Seed":
            seed,

        "Best_C":
            svm_search.best_params_[
                "classifier__C"
            ],

        "Best_kernel":
            svm_search.best_params_[
                "classifier__kernel"
            ],

        "Best_CV_ROC_AUC":
            svm_search.best_score_
    })

    print("Best parameters:")

    print(
        svm_search.best_params_
    )

    print(
        f"Best 5-fold CV ROC-AUC: "
        f"{svm_search.best_score_:.4f}"
    )

    # --------------------------------------------------------------
    # Independent-test predictions
    # --------------------------------------------------------------

    y_probability = (
        best_model.predict_proba(
            X_test_49
        )[:, 1]
    )

    y_prediction = (
        best_model.predict(
            X_test_49
        )
    )

    svm_seed_probabilities[
        seed
    ] = y_probability

    svm_seed_predictions[
        seed
    ] = y_prediction

    # --------------------------------------------------------------
    # Independent-test metrics
    # --------------------------------------------------------------

    roc_auc = roc_auc_score(
        y_test_49,
        y_probability
    )

    pr_auc = average_precision_score(
        y_test_49,
        y_probability
    )

    f1 = f1_score(
        y_test_49,
        y_prediction
    )

    precision = precision_score(
        y_test_49,
        y_prediction,
        zero_division=0
    )

    recall = recall_score(
        y_test_49,
        y_prediction,
        zero_division=0
    )

    accuracy = accuracy_score(
        y_test_49,
        y_prediction
    )

    tn, fp, fn, tp = (
        confusion_matrix(
            y_test_49,
            y_prediction
        )
        .ravel()
    )

    specificity = (
        tn /
        (tn + fp)
    )

    # --------------------------------------------------------------
    # Store results
    # --------------------------------------------------------------

    svm_seed_results.append({

        "Seed":
            seed,

        "ROC_AUC":
            roc_auc,

        "PR_AUC":
            pr_auc,

        "F1":
            f1,

        "Precision":
            precision,

        "Recall":
            recall,

        "Specificity":
            specificity,

        "Accuracy":
            accuracy,

        "TN":
            int(tn),

        "FP":
            int(fp),

        "FN":
            int(fn),

        "TP":
            int(tp)
    })

    print()

    print(
        f"Independent-test ROC-AUC: "
        f"{roc_auc:.4f}"
    )

    print(
        f"Independent-test PR-AUC:  "
        f"{pr_auc:.4f}"
    )

    print(
        f"Independent-test F1:      "
        f"{f1:.4f}"
    )

# ======================================================================
# 5. CREATE RESULT TABLES
# ======================================================================

svm_seed_results_df = pd.DataFrame(
    svm_seed_results
)

svm_best_params_df = pd.DataFrame(
    svm_best_params
)

assert len(
    svm_seed_results_df
) == 3

assert len(
    svm_best_params_df
) == 3

assert set(
    svm_seed_results_df["Seed"]
) == {
    42,
    123,
    7
}

assert set(
    svm_best_params_df["Seed"]
) == {
    42,
    123,
    7
}

# ======================================================================
# 6. ROBUSTNESS SUMMARY
# ======================================================================

SVM_METRICS = [
    "ROC_AUC",
    "PR_AUC",
    "F1",
    "Precision",
    "Recall",
    "Specificity",
    "Accuracy"
]

assert np.isfinite(
    svm_seed_results_df[
        SVM_METRICS
    ].to_numpy()
).all()

svm_robustness_rows = []

for metric in SVM_METRICS:

    values = (
        svm_seed_results_df[
            metric
        ]
        .astype(float)
    )

    mean_value = values.mean()

    sd_value = values.std(
        ddof=1
    )

    svm_robustness_rows.append({

        "Metric":
            metric,

        "Mean":
            mean_value,

        "SD":
            sd_value,

        "Minimum":
            values.min(),

        "Maximum":
            values.max(),

        "Range":
            (
                values.max()
                -
                values.min()
            ),

        "CV_Percent":
            (
                sd_value /
                mean_value
            ) * 100
    })

svm_robustness_summary_df = pd.DataFrame(
    svm_robustness_rows
)

assert len(
    svm_robustness_summary_df
) == 7

assert np.isfinite(
    svm_robustness_summary_df[
        [
            "Mean",
            "SD",
            "Minimum",
            "Maximum",
            "Range",
            "CV_Percent"
        ]
    ].to_numpy()
).all()

# ======================================================================
# 7. DISPLAY RESULTS
# ======================================================================

print()
print("=" * 70)
print("THREE-SEED SUPPORT VECTOR MACHINE TEST RESULTS")
print("=" * 70)

print(
    svm_seed_results_df.to_string(
        index=False,
        formatters={
            metric: "{:.4f}".format
            for metric in SVM_METRICS
        }
    )
)

print()
print("=" * 70)
print("THREE-SEED SUPPORT VECTOR MACHINE MEAN ± SD")
print("=" * 70)

for _, row in (
    svm_robustness_summary_df
    .iterrows()
):

    print(
        f"{row['Metric']:<15}: "
        f"{row['Mean']:.4f} ± "
        f"{row['SD']:.4f}"
    )

# ======================================================================
# 8. SAVE NOTEBOOK 14 ARTIFACTS
# ======================================================================

NOTEBOOK14_DIR = (
    RESULTS_DIR /
    "notebook14_fairness"
)

NOTEBOOK14_DIR.mkdir(
    parents=True,
    exist_ok=True
)

SVM_RESULTS_PATH = (
    NOTEBOOK14_DIR /
    "svm_three_seed_test_results.csv"
)

SVM_SUMMARY_PATH = (
    NOTEBOOK14_DIR /
    "svm_three_seed_robustness_summary.csv"
)

SVM_PARAMS_PATH = (
    NOTEBOOK14_DIR /
    "svm_three_seed_best_params.csv"
)

svm_seed_results_df.to_csv(
    SVM_RESULTS_PATH,
    index=False
)

svm_robustness_summary_df.to_csv(
    SVM_SUMMARY_PATH,
    index=False
)

svm_best_params_df.to_csv(
    SVM_PARAMS_PATH,
    index=False
)

assert SVM_RESULTS_PATH.exists()
assert SVM_SUMMARY_PATH.exists()
assert SVM_PARAMS_PATH.exists()

# ======================================================================
# 9. FINAL VALIDATION
# ======================================================================

print()
print("=" * 70)
print("CELL 8 VALIDATION")
print("=" * 70)

print(
    "✓ Support Vector Machine evaluated across three seeds"
)

print(
    "✓ Seed-specific 5-fold tuning used"
)

print(
    "✓ Original Notebook 9 SVM search space preserved"
)

print(
    f"✓ {svm_candidate_count} SVM candidate "
    "configurations evaluated per seed"
)

print(
    "✓ StandardScaler fitted inside each CV pipeline"
)

print(
    "✓ Matched 49-feature predictor space used"
)

print(
    "✓ Same 468-observation training set used"
)

print(
    "✓ Same 117-observation independent test set used"
)

print(
    "✓ ROC-AUC used for tuning"
)

print(
    "✓ Independent test set not used during tuning"
)

print(
    "✓ Seven independent-test metrics calculated"
)

print(
    "✓ Confusion matrices calculated"
)

print(
    "✓ Three-seed mean ± SD calculated"
)

print(
    "✓ Best parameters saved per seed"
)

print(
    "✓ Support Vector Machine robustness artifacts saved"
)

print(
    f"✓ {SVM_RESULTS_PATH}"
)

print(
    f"✓ {SVM_SUMMARY_PATH}"
)

print(
    f"✓ {SVM_PARAMS_PATH}"
)

print(
    "✓ No feature selection rerun"
)

print(
    "✓ No previous notebook artifact modified"
)

print(
    "✓ Notebooks 11–13 remain locked"
)

print()
print("=" * 70)
print("✓ CELL 8 — COMPLETED")
print("=" * 70)
```
