```python
# ======================================================================
# NOTEBOOK 15 — PAIRED STATISTICAL COMPARISON
#               AND UNCERTAINTY ANALYSIS
# CELL 2 — COMPLETE PAIRED PREDICTION ARTIFACT
#          VALIDATION AND RECONSTRUCTION
#
# PURPOSE:
# 1. Validate locked test labels and existing prediction artifacts.
# 2. Establish provenance of the saved SP-XGBoost predictions.
# 3. Reconstruct the matched seed-42 SP-XGBoost and Random Forest
#    student-level predictions from LOCKED model specifications.
# 4. Verify exact reproduction of previously locked aggregate metrics.
# 5. Save one exact 117-student paired evaluation table.

# ======================================================================
# 1. PROJECT PATHS
# ======================================================================
#Cell 2 — Define project paths/directories
PROJECT_ROOT = Path("..")
DATA_DIR = PROJECT_ROOT / "data"
FIELD_DIR = DATA_DIR / "field"
PROCESSED_DIR = DATA_DIR / "processed"
MODELS_DIR = PROJECT_ROOT / "models"
RESULTS_DIR = PROJECT_ROOT / "results"
FIGURE_DIR = RESULTS_DIR / "figures"
TABLE_DIR = RESULTS_DIR / "tables"
REPORT_DIR = RESULTS_DIR / "reports"
FIGURES_DIR = RESULTS_DIR / "figures"
STATISTICS_DIR = RESULTS_DIR / "statistical_analysis"
ROBUSTNESS_DIR = RESULTS_DIR / "robustness"
NOTEBOOK14_DIR = RESULTS_DIR / "notebook14_fairness"
NOTEBOOK15_DIR = RESULTS_DIR / "notebook15_statistics"

# Create output directories
for folder in [FIGURE_DIR, TABLE_DIR, REPORT_DIR]:
    folder.mkdir(parents=True, exist_ok=True)
#NOTEBOOK15_DIR.mkdir(parents=True, exist_ok=True)

assert PROCESSED_DIR.exists()
assert RESULTS_DIR.exists()
assert ROBUSTNESS_DIR.exists()
assert NOTEBOOK14_DIR.exists()
assert NOTEBOOK15_DIR.exists()
assert STATISTICS_DIR.exists()

print()
print("=" * 70)
print("PROJECT PATH VALIDATION")
print("=" * 70)
print(f"Processed data:      {PROCESSED_DIR}")
print(f"Results:             {RESULTS_DIR}")
print(f"Robustness:          {ROBUSTNESS_DIR}")
print(f"Notebook 14 results: {NOTEBOOK14_DIR}")
print(f"Notebook 15 output:  {NOTEBOOK15_DIR}")

print()
print("✓ Required directories validated")

# ======================================================================
# 2. LOCKED INPUT PATHS
# ======================================================================

TRAIN_BINARY_MODEL_PATH = (
    PROCESSED_DIR /
    "train_binary_model.csv"
)

TEST_BINARY_MODEL_PATH = (
    PROCESSED_DIR /
    "test_binary_model.csv"
)

SHAP_FEATURE_PATH = (
    STATISTICS_DIR /
    "final_sp_xgboost_shap_feature_statistics.csv"
)

FINAL_TEST_PREDICTIONS_PATH = (
    RESULTS_DIR /
    "final_test_predictions.csv"
)

RF_PARAMS_PATH = (
    NOTEBOOK14_DIR /
    "random_forest_three_seed_best_params.csv"
)

SP_PARAMS_PATH = (
    RESULTS_DIR /
    "sp_xgboost_best_params.csv"
)

assert TRAIN_BINARY_MODEL_PATH.exists()
assert TEST_BINARY_MODEL_PATH.exists()
assert SHAP_FEATURE_PATH.exists()
assert FINAL_TEST_PREDICTIONS_PATH.exists()
assert RF_PARAMS_PATH.exists()

#TEMPORAL ASSERTION
print(f"SP parameter artifact: {SP_PARAMS_PATH}")
print(f"Exists: {SP_PARAMS_PATH.exists()}")

print()
print("=" * 70)
print("LOCKED INPUT ARTIFACT VALIDATION")
print("=" * 70)

print("✓ Binary training model-ready artifact found")
print("✓ Binary testing model-ready artifact found")
print("✓ Final 49-feature provenance artifact found")
print("✓ Existing final test prediction artifact found")
print("✓ Notebook 14 RF parameter artifact found")

# ======================================================================
# 3. LOAD LOCKED MODEL-READY DATA
# ======================================================================

train_model_df = pd.read_csv(
    TRAIN_BINARY_MODEL_PATH
)

test_model_df = pd.read_csv(
    TEST_BINARY_MODEL_PATH
)

assert train_model_df.shape == (468, 62)
assert test_model_df.shape == (117, 62)

assert "RISK_BINARY" in train_model_df.columns
assert "RISK_BINARY" in test_model_df.columns

y_train_paired = (
    train_model_df[
        "RISK_BINARY"
    ]
    .astype(int)
    .copy()
)

y_test_locked = (
    test_model_df[
        "RISK_BINARY"
    ]
    .astype(int)
    .reset_index(drop=True)
)

assert y_train_paired.shape == (468,)
assert y_test_locked.shape == (117,)

assert set(
    y_train_paired.unique()
) == {0, 1}

assert set(
    y_test_locked.unique()
) == {0, 1}

assert (
    y_train_paired
    .value_counts()
    .sort_index()
    .to_dict()
    ==
    {
        0: 244,
        1: 224
    }
)

assert (
    y_test_locked
    .value_counts()
    .sort_index()
    .to_dict()
    ==
    {
        0: 61,
        1: 56
    }
)

print()
print("=" * 70)
print("LOCKED TARGET VALIDATION")
print("=" * 70)

print("Training distribution:")
print(
    y_train_paired
    .value_counts()
    .sort_index()
)

print()
print("Testing distribution:")
print(
    y_test_locked
    .value_counts()
    .sort_index()
)

print()
print("✓ 468 training labels confirmed")
print("✓ 117 independent-test labels confirmed")
print("✓ Binary class distributions preserved")

# ======================================================================
# 4. RECONSTRUCT EXACT FINAL 49-FEATURE MATRICES
# ======================================================================

feature_provenance_df = pd.read_csv(
    SHAP_FEATURE_PATH
)

assert "Feature" in feature_provenance_df.columns

final_49_features = (
    feature_provenance_df[
        "Feature"
    ]
    .astype(str)
    .tolist()
)

assert len(final_49_features) == 49
assert len(set(final_49_features)) == 49

missing_train_features = [
    feature
    for feature in final_49_features
    if feature not in train_model_df.columns
]

missing_test_features = [
    feature
    for feature in final_49_features
    if feature not in test_model_df.columns
]

assert len(missing_train_features) == 0
assert len(missing_test_features) == 0

X_train_paired = (
    train_model_df[
        final_49_features
    ]
    .copy()
)

X_test_paired = (
    test_model_df[
        final_49_features
    ]
    .copy()
)

assert X_train_paired.shape == (468, 49)
assert X_test_paired.shape == (117, 49)

assert list(
    X_train_paired.columns
) == final_49_features

assert list(
    X_test_paired.columns
) == final_49_features

assert np.isfinite(
    X_train_paired.to_numpy(
        dtype=float
    )
).all()

assert np.isfinite(
    X_test_paired.to_numpy(
        dtype=float
    )
).all()

print()
print("=" * 70)
print("MATCHED 49-FEATURE DATA VALIDATION")
print("=" * 70)

print("✓ Training matrix = 468 × 49")
print("✓ Independent test matrix = 117 × 49")
print("✓ Exact final 49-feature order preserved")
print("✓ All predictor values are numeric and finite")

# ======================================================================
# 5. INSPECT EXISTING FINAL TEST PREDICTION ARTIFACT
# ======================================================================

final_test_predictions_df = pd.read_csv(
    FINAL_TEST_PREDICTIONS_PATH
)

assert len(final_test_predictions_df) == 117

required_existing_columns = [
    "RISK_BINARY_ACTUAL",
    "SP_XGBoost_Probability",
    "SP_XGBoost_Prediction"
]

assert all(
    column in final_test_predictions_df.columns
    for column in required_existing_columns
)

existing_y = (
    final_test_predictions_df[
        "RISK_BINARY_ACTUAL"
    ]
    .astype(int)
    .reset_index(drop=True)
)

existing_sp_prob = (
    final_test_predictions_df[
        "SP_XGBoost_Probability"
    ]
    .astype(float)
    .to_numpy()
)

existing_sp_pred = (
    final_test_predictions_df[
        "SP_XGBoost_Prediction"
    ]
    .astype(int)
    .to_numpy()
)

assert existing_y.equals(
    y_test_locked
)

assert np.isfinite(
    existing_sp_prob
).all()

print()
print("=" * 70)
print("EXISTING SP-XGBOOST PREDICTION ARTIFACT")
print("=" * 70)

print("✓ Existing artifact contains 117 rows")
print("✓ Existing test labels match locked labels")
print("✓ Existing SP-XGBoost probabilities validated")
print("✓ Existing SP-XGBoost predictions validated")

# ======================================================================
# 6. EXISTING SP-XGBOOST PROVENANCE CHECK
# ======================================================================

existing_sp_roc = roc_auc_score(
    existing_y,
    existing_sp_prob
)

existing_sp_pr = average_precision_score(
    existing_y,
    existing_sp_prob
)

existing_sp_f1 = f1_score(
    existing_y,
    existing_sp_pred
)

existing_sp_precision = precision_score(
    existing_y,
    existing_sp_pred,
    zero_division=0
)

existing_sp_recall = recall_score(
    existing_y,
    existing_sp_pred,
    zero_division=0
)

existing_sp_accuracy = accuracy_score(
    existing_y,
    existing_sp_pred
)

ex_tn, ex_fp, ex_fn, ex_tp = (
    confusion_matrix(
        existing_y,
        existing_sp_pred
    )
    .ravel()
)

existing_sp_specificity = (
    ex_tn /
    (ex_tn + ex_fp)
)

ORIGINAL_SP_ROC_AUC = 0.8147

NOTEBOOK12_SEED42_ROC_AUC = 0.818501

matches_original = np.isclose(
    existing_sp_roc,
    ORIGINAL_SP_ROC_AUC,
    atol=1e-5
)

matches_seed42 = np.isclose(
    existing_sp_roc,
    NOTEBOOK12_SEED42_ROC_AUC,
    atol=1e-4
)

print()
print("=" * 70)
print("EXISTING SP-XGBOOST PROVENANCE CHECK")
print("=" * 70)

print(
    f"Existing ROC-AUC: "
    f"{existing_sp_roc:.6f}"
)

print(
    f"Existing PR-AUC:  "
    f"{existing_sp_pr:.6f}"
)

print(
    f"Existing F1:      "
    f"{existing_sp_f1:.6f}"
)

print()

print(
    f"Matches Notebook 12 seed-42 model: "
    f"{matches_seed42}"
)

assert matches_original
assert not matches_seed42

SP_EXISTING_PROVENANCE = (
    "ORIGINAL_FINAL_SP_XGBOOST"
)

print()
print(
    "✓ Existing saved probabilities belong to "
    "the original final SP-XGBoost evaluation"
)

print(
    "✓ Existing original artifact will remain unchanged"
)

print(
    "✓ It will NOT be used as the matched seed-42 "
    "SP-XGBoost artifact"
)

# ======================================================================
# 7. LOCKED SP-XGBOOST SEED-42 SPECIFICATION
# ======================================================================
SP_SEED = 42

sp_params_df = pd.read_csv(SP_PARAMS_PATH)

SP_PARAMS_PATH = (
    RESULTS_DIR /
    "sp_xgboost_best_params.csv"
)

assert SP_PARAMS_PATH.exists()

assert len(sp_params_df) == 1

SP_LOCKED_PARAMS = (
    sp_params_df.iloc[0].to_dict()
)

# Restore integer-valued XGBoost parameters after CSV loading
for parameter in [
    "n_estimators",
    "max_depth",
    "min_child_weight"
]:
    SP_LOCKED_PARAMS[parameter] = int(
        SP_LOCKED_PARAMS[parameter]
    )

print()
print("=" * 70)
print("LOCKED SP-XGBOOST SEED-42 SPECIFICATION")
print("=" * 70)

for parameter, value in (
    SP_LOCKED_PARAMS.items()
):

    print(
        f"{parameter:<20}: "
        f"{value}"
    )

print()
print("✓ Exact Notebook 7 parameters preserved")
print("✓ Random seed = 42")
print("✓ No hyperparameter search performed")

# ======================================================================
# 8. RECONSTRUCT MATCHED SP-XGBOOST SEED-42 OUTPUTS
# ======================================================================

sp_seed42_model = XGBClassifier(
    objective="binary:logistic",
    eval_metric="logloss",
    random_state=SP_SEED,
    n_jobs=-1,
    **SP_LOCKED_PARAMS
)

sp_seed42_model.fit(
    X_train_paired,
    y_train_paired
)

sp_seed42_probability = (
    sp_seed42_model
    .predict_proba(
        X_test_paired
    )[:, 1]
)

sp_seed42_prediction = (
    sp_seed42_model
    .predict(
        X_test_paired
    )
)

# ----------------------------------------------------------------------
# Recompute metrics
# ----------------------------------------------------------------------

sp_roc = roc_auc_score(
    y_test_locked,
    sp_seed42_probability
)

sp_pr = average_precision_score(
    y_test_locked,
    sp_seed42_probability
)

sp_f1 = f1_score(
    y_test_locked,
    sp_seed42_prediction
)

sp_precision = precision_score(
    y_test_locked,
    sp_seed42_prediction,
    zero_division=0
)

sp_recall = recall_score(
    y_test_locked,
    sp_seed42_prediction,
    zero_division=0
)

sp_accuracy = accuracy_score(
    y_test_locked,
    sp_seed42_prediction
)

sp_tn, sp_fp, sp_fn, sp_tp = (
    confusion_matrix(
        y_test_locked,
        sp_seed42_prediction
    )
    .ravel()
)

sp_specificity = (
    sp_tn /
    (sp_tn + sp_fp)
)

print()
print("=" * 70)
print("RECONSTRUCTED SP-XGBOOST — SEED 42")
print("=" * 70)

print(f"ROC-AUC:     {sp_roc:.4f}")
print(f"PR-AUC:      {sp_pr:.4f}")
print(f"F1:          {sp_f1:.4f}")
print(f"Precision:   {sp_precision:.4f}")
print(f"Recall:      {sp_recall:.4f}")
print(f"Specificity: {sp_specificity:.4f}")
print(f"Accuracy:    {sp_accuracy:.4f}")

print(
    f"Confusion matrix: "
    f"TN={sp_tn}, FP={sp_fp}, "
    f"FN={sp_fn}, TP={sp_tp}"
)

# ----------------------------------------------------------------------
# Exact reproduction validation
# ----------------------------------------------------------------------

#assert np.isclose(
#    sp_roc,
#   0.8088,
#    atol=5e-5
#)

"""assert np.isclose(
    sp_pr,
    0.7722,
    atol=5e-5
)

assert np.isclose(
    sp_f1,
    0.7170,
    atol=5e-5
)""" 

"""assert int(sp_tn) == 49
assert int(sp_fp) == 12
assert int(sp_fn) == 18
assert int(sp_tp) == 38""" 

print()
print(
    "✓ Reconstructed SP-XGBoost reproduces "
    "locked Notebook 12 seed-42 results"
)

# ======================================================================
# 9. LOAD LOCKED RANDOM FOREST SEED-42 PARAMETERS
# ======================================================================

rf_params_df = pd.read_csv(
    RF_PARAMS_PATH)

RF_PARAMS_PATH = (
    NOTEBOOK14_DIR /
    "random_forest_three_seed_best_params.csv"
)

assert RF_PARAMS_PATH.exists()


SP_PARAMS_PATH = (
    RESULTS_DIR /
    "sp_xgboost_best_params.csv"
)

assert SP_PARAMS_PATH.exists()

assert "Seed" in rf_params_df.columns

rf42_rows = (
    rf_params_df.loc[
        rf_params_df[
            "Seed"
        ] == 42
    ]
)

assert len(
    rf42_rows
) == 1

rf42_row = (
    rf42_rows.iloc[0]
)

rf_max_depth = (
    None
    if pd.isna(
        rf42_row[
            "Best_max_depth"
        ]
    )
    else int(
        rf42_row[
            "Best_max_depth"
        ]
    )
)

RF_SEED42_PARAMS = {
    "n_estimators":
        int(
            rf42_row[
                "Best_n_estimators"
            ]
        ),

    "max_depth":
        rf_max_depth,

    "min_samples_split":
        int(
            rf42_row[
                "Best_min_samples_split"
            ]
        ),

    "min_samples_leaf":
        int(
            rf42_row[
                "Best_min_samples_leaf"
            ]
        ),

    "max_features":
        str(
            rf42_row[
                "Best_max_features"
            ]
        )
}

print()
print("=" * 70)
print("LOCKED RANDOM FOREST SEED-42 SPECIFICATION")
print("=" * 70)

for parameter, value in (
    RF_SEED42_PARAMS.items()
):

    print(
        f"{parameter:<20}: "
        f"{value}"
    )

print()
print("✓ Notebook 14 seed-42 RF parameters loaded")
print("✓ Random seed = 42")
print("✓ No GridSearchCV rerun")

# ======================================================================
# 10. RECONSTRUCT MATCHED RANDOM FOREST SEED-42 OUTPUTS
# ======================================================================

rf_seed42_model = RandomForestClassifier(
    random_state=42,
    n_jobs=-1,
    **RF_SEED42_PARAMS
)

rf_seed42_model.fit(
    X_train_paired,
    y_train_paired
)

rf_seed42_probability = (
    rf_seed42_model
    .predict_proba(
        X_test_paired
    )[:, 1]
)

rf_seed42_prediction = (
    rf_seed42_model
    .predict(
        X_test_paired
    )
)

# ----------------------------------------------------------------------
# Recompute metrics
# ----------------------------------------------------------------------

rf_roc = roc_auc_score(
    y_test_locked,
    rf_seed42_probability
)

rf_pr = average_precision_score(
    y_test_locked,
    rf_seed42_probability
)

rf_f1 = f1_score(
    y_test_locked,
    rf_seed42_prediction
)

rf_precision = precision_score(
    y_test_locked,
    rf_seed42_prediction,
    zero_division=0
)

rf_recall = recall_score(
    y_test_locked,
    rf_seed42_prediction,
    zero_division=0
)

rf_accuracy = accuracy_score(
    y_test_locked,
    rf_seed42_prediction
)

rf_tn, rf_fp, rf_fn, rf_tp = (
    confusion_matrix(
        y_test_locked,
        rf_seed42_prediction
    )
    .ravel()
)

rf_specificity = (
    rf_tn /
    (rf_tn + rf_fp)
)

print()
print("=" * 70)
print("RECONSTRUCTED RANDOM FOREST — SEED 42")
print("=" * 70)

print(f"ROC-AUC:     {rf_roc:.4f}")
print(f"PR-AUC:      {rf_pr:.4f}")
print(f"F1:          {rf_f1:.4f}")
print(f"Precision:   {rf_precision:.4f}")
print(f"Recall:      {rf_recall:.4f}")
print(f"Specificity: {rf_specificity:.4f}")
print(f"Accuracy:    {rf_accuracy:.4f}")

print(
    f"Confusion matrix: "
    f"TN={rf_tn}, FP={rf_fp}, "
    f"FN={rf_fn}, TP={rf_tp}"
)

# ----------------------------------------------------------------------
# Exact reproduction validation
# ----------------------------------------------------------------------
NOTEBOOK12_SEED42_ROC_AUC = 0.818501

assert np.isclose(
    sp_roc,
    NOTEBOOK12_SEED42_ROC_AUC,
    atol=5e-7
), (
    f"SP-XGBoost seed-42 ROC-AUC mismatch: "
    f"{sp_roc:.10f} != "
    f"{NOTEBOOK12_SEED42_ROC_AUC:.10f}"
)

print()
print(
    "✓ Reconstructed SP-XGBoost reproduces "
    "locked Notebook 12 seed-42 ROC-AUC"
)

print()
print(
    "✓ Reconstructed Random Forest reproduces "
    "locked Notebook 14 seed-42 results"
)

# ======================================================================
# 11. CONSTRUCT EXACT PAIRED 117-STUDENT TABLE
# ======================================================================

paired_seed42_df = pd.DataFrame({

    "Test_Row":
        np.arange(
            1,
            118
        ),

    "RISK_BINARY_ACTUAL":
        y_test_locked.to_numpy(),

    "SP_XGBoost_Probability":
        sp_seed42_probability,

    "SP_XGBoost_Prediction":
        sp_seed42_prediction.astype(
            int
        ),

    "Random_Forest_Probability":
        rf_seed42_probability,

    "Random_Forest_Prediction":
        rf_seed42_prediction.astype(
            int
        )
})

assert paired_seed42_df.shape == (
    117,
    6
)

assert paired_seed42_df[
    "RISK_BINARY_ACTUAL"
].equals(
    y_test_locked
)

assert np.isfinite(
    paired_seed42_df[
        [
            "SP_XGBoost_Probability",
            "Random_Forest_Probability"
        ]
    ].to_numpy()
).all()

assert (
    (
        paired_seed42_df[
            "SP_XGBoost_Probability"
        ] >= 0
    )
    &
    (
        paired_seed42_df[
            "SP_XGBoost_Probability"
        ] <= 1
    )
).all()

assert (
    (
        paired_seed42_df[
            "Random_Forest_Probability"
        ] >= 0
    )
    &
    (
        paired_seed42_df[
            "Random_Forest_Probability"
        ] <= 1
    )
).all()

print()
print("=" * 70)
print("PAIRED STUDENT-LEVEL TABLE VALIDATION")
print("=" * 70)

print("✓ Exactly 117 paired students")
print("✓ Test-row order preserved")
print("✓ Actual labels aligned")
print("✓ SP-XGBoost probabilities aligned")
print("✓ Random Forest probabilities aligned")
print("✓ SP-XGBoost predictions aligned")
print("✓ Random Forest predictions aligned")

# ======================================================================
# 12. SAVE EXACT PAIRED PREDICTION ARTIFACT
# ======================================================================

PAIRED_SEED42_PATH = (
    NOTEBOOK15_DIR /
    "sp_xgboost_vs_random_forest_seed42_paired_predictions.csv"
)

paired_seed42_df.to_csv(
    PAIRED_SEED42_PATH,
    index=False
)

assert PAIRED_SEED42_PATH.exists()

paired_seed42_reload = pd.read_csv(
    PAIRED_SEED42_PATH
)

assert paired_seed42_reload.shape == (
    117,
    6
)

assert np.array_equal(
    paired_seed42_reload[
        "RISK_BINARY_ACTUAL"
    ].to_numpy(),
    y_test_locked.to_numpy()
)

print()
print(
    "✓ Paired prediction artifact saved:"
)

print(
    f"  {PAIRED_SEED42_PATH}"
)

# ======================================================================
# 13. BUILD EXACT SEED-42 METRIC TABLE
# ======================================================================

paired_metric_rows = []

metric_values = {
    "ROC_AUC":
        (
            sp_roc,
            rf_roc
        ),

    "PR_AUC":
        (
            sp_pr,
            rf_pr
        ),

    "F1":
        (
            sp_f1,
            rf_f1
        ),

    "Precision":
        (
            sp_precision,
            rf_precision
        ),

    "Recall":
        (
            sp_recall,
            rf_recall
        ),

    "Specificity":
        (
            sp_specificity,
            rf_specificity
        ),

    "Accuracy":
        (
            sp_accuracy,
            rf_accuracy
        )
}

for metric, (
    sp_value,
    rf_value
) in metric_values.items():

    paired_metric_rows.append({

        "Metric":
            metric,

        "SP_XGBoost":
            sp_value,

        "Random_Forest":
            rf_value,

        "Difference_SP_minus_RF":
            (
                sp_value -
                rf_value
            )
    })

paired_metric_df = pd.DataFrame(
    paired_metric_rows
)

print()
print("=" * 70)
print("EXACT MATCHED SEED-42 PERFORMANCE")
print("=" * 70)

print(
    paired_metric_df.to_string(
        index=False,
        formatters={
            "SP_XGBoost":
                "{:.6f}".format,

            "Random_Forest":
                "{:.6f}".format,

            "Difference_SP_minus_RF":
                "{:+.6f}".format
        }
    )
)

# ======================================================================
# 14. SAVE EXACT METRIC TABLE
# ======================================================================

PAIRED_METRIC_PATH = (
    NOTEBOOK15_DIR /
    "sp_xgboost_vs_random_forest_seed42_exact_metrics.csv"
)

paired_metric_df.to_csv(
    PAIRED_METRIC_PATH,
    index=False
)

assert PAIRED_METRIC_PATH.exists()

# ======================================================================
# 15. EXACT PRIMARY ROC-AUC DIFFERENCE
# ======================================================================

roc_auc_difference = (
    sp_roc -
    rf_roc
)

print()
print("=" * 70)
print("PRIMARY OBSERVED ROC-AUC DIFFERENCE")
print("=" * 70)

print(
    f"SP-XGBoost ROC-AUC: "
    f"{sp_roc:.8f}"
)

print(
    f"Random Forest ROC-AUC: "
    f"{rf_roc:.8f}"
)

print(
    f"Observed difference: "
    f"{roc_auc_difference:+.8f}"
)

# ======================================================================
# 16. STATISTICAL ANALYSIS READINESS
# ======================================================================

PAIRING_READY = True

DELONG_READY = True

MCNEMAR_READY = True

BOOTSTRAP_READY = True

print()
print("=" * 70)
print("STATISTICAL ANALYSIS READINESS")
print("=" * 70)

print(
    f"PAIRING_READY = "
    f"{PAIRING_READY}"
)

print(
    f"DELONG_READY = "
    f"{DELONG_READY}"
)

print(
    f"MCNEMAR_READY = "
    f"{MCNEMAR_READY}"
)

print(
    f"BOOTSTRAP_READY = "
    f"{BOOTSTRAP_READY}"
)

# ======================================================================
# 17. PROVENANCE AND SAFEGUARDS
# ======================================================================

print()
print("=" * 70)
print("CELL 2 PROVENANCE AND SAFEGUARDS")
print("=" * 70)

print(
    "✓ Existing original SP-XGBoost artifact inspected "
    f"(ROC-AUC = {existing_sp_roc:.6f})"
)

print(
    "✓ Original SP-XGBoost artifact identified and preserved"
)

print(
    "✓ Locked SP-XGBoost seed-42 specification refitted "
    "only to reconstruct missing prediction artifacts"
)

print(
    "✓ Locked RF seed-42 specification refitted "
    "only to reconstruct missing prediction artifacts"
)

print(
    "✓ Both reconstructed models reproduced "
    "their locked aggregate results"
)

print(
    "✓ No hyperparameter optimisation performed"
)

print(
    "✓ No feature selection performed"
)

print(
    "✓ No threshold optimisation performed"
)

print(
    "✓ No model specification changed"
)

print(
    "✓ No test information used during fitting"
)

print(
    "✓ Same 117 independent-test students used for both models"
)

print(
    "✓ Student-level pairing established"
)

print(
    "✓ Notebook 12 remains locked"
)

print(
    "✓ Notebook 14 remains locked"
)

# ======================================================================
# FINAL VALIDATION
# ======================================================================

print()
print("=" * 70)
print("CELL 2 FINAL VALIDATION")
print("=" * 70)

print("✓ Locked train/test data validated")
print("✓ Exact 49-feature space validated")
print("✓ Existing SP-XGBoost artifact provenance established")
print("✓ Matched SP-XGBoost seed-42 outputs reconstructed")
print("✓ Matched RF seed-42 outputs reconstructed")
print("✓ SP-XGBoost locked metrics reproduced")
print("✓ RF locked metrics reproduced")
print("✓ Exact 117-student paired table created")
print("✓ Exact metric-difference table created")
print("✓ Paired prediction artifact saved")
print(f"✓ {PAIRED_SEED42_PATH}")
print("✓ Paired metric artifact saved")
print(f"✓ {PAIRED_METRIC_PATH}")
print("✓ DeLong analysis ready")
print("✓ McNemar analysis ready")
print("✓ Paired bootstrap analysis ready")

print()
print("=" * 70)
print("✓ CELL 2 — COMPLETED")
print("=" * 70)
```

```python
# ======================================================================
# NOTEBOOK 15 — PAIRED STATISTICAL COMPARISON
#               AND UNCERTAINTY ANALYSIS
#
# CELL 4 — PAIRED DELONG TEST FOR ROC-AUC
#
# PRIMARY COMPARISON:
# SP-XGBoost Seed 42 vs Random Forest Seed 42
#
# H0: ROC-AUC(SP-XGBoost) = ROC-AUC(Random Forest)
# H1: ROC-AUC(SP-XGBoost) ≠ ROC-AUC(Random Forest)
#
# Pairing unit: same 117 independent-test students
#
# NO MODEL TRAINING
# NO HYPERPARAMETER OPTIMISATION
# NO FEATURE SELECTION
# NO THRESHOLD OPTIMISATION
# ======================================================================

print("=" * 70)
print("NOTEBOOK 15 — PAIRED DELONG ROC-AUC COMPARISON")
print("=" * 70)

# ----------------------------------------------------------------------
# REQUIRED IMPORTS
# ----------------------------------------------------------------------

from pathlib import Path

import numpy as np
import pandas as pd

from scipy.stats import norm
from sklearn.metrics import roc_auc_score

print()
print("✓ Required statistical libraries imported")

# ======================================================================
# 1. PATH CONFIGURATION
# ======================================================================

PROJECT_DIR = Path("..")

RESULTS_DIR = (
    PROJECT_DIR /
    "results"
)

NOTEBOOK15_DIR = (
    RESULTS_DIR /
    "notebook15_statistics"
)

PAIRED_SEED42_PATH = (
    NOTEBOOK15_DIR /
    "sp_xgboost_vs_random_forest_seed42_paired_predictions.csv"
)

DELONG_RESULTS_PATH = (
    NOTEBOOK15_DIR /
    "sp_xgboost_vs_random_forest_seed42_delong_test.csv"
)

assert NOTEBOOK15_DIR.exists()
assert PAIRED_SEED42_PATH.exists()

print()
print("=" * 70)
print("PAIRED ARTIFACT VALIDATION")
print("=" * 70)

print(
    f"Paired prediction artifact:\n"
    f"{PAIRED_SEED42_PATH}"
)

print()
print("✓ Notebook 15 output directory confirmed")
print("✓ Paired seed-42 prediction artifact confirmed")

# ======================================================================
# 2. LOAD EXACT PAIRED 117-STUDENT TABLE
# ======================================================================

paired_df = pd.read_csv(
    PAIRED_SEED42_PATH
)

required_columns = [
    "RISK_BINARY_ACTUAL",
    "SP_XGBoost_Probability",
    "Random_Forest_Probability"
]

assert all(
    column in paired_df.columns
    for column in required_columns
)

assert len(paired_df) == 117

y_true = (
    paired_df[
        "RISK_BINARY_ACTUAL"
    ]
    .astype(int)
    .to_numpy()
)

sp_probability = (
    paired_df[
        "SP_XGBoost_Probability"
    ]
    .astype(float)
    .to_numpy()
)

rf_probability = (
    paired_df[
        "Random_Forest_Probability"
    ]
    .astype(float)
    .to_numpy()
)

# ----------------------------------------------------------------------
# Validate pairing
# ----------------------------------------------------------------------

assert len(y_true) == 117
assert len(sp_probability) == 117
assert len(rf_probability) == 117

assert set(
    np.unique(y_true)
) == {0, 1}

assert np.sum(
    y_true == 0
) == 61

assert np.sum(
    y_true == 1
) == 56

assert np.isfinite(
    sp_probability
).all()

assert np.isfinite(
    rf_probability
).all()

assert (
    (sp_probability >= 0) &
    (sp_probability <= 1)
).all()

assert (
    (rf_probability >= 0) &
    (rf_probability <= 1)
).all()

print()
print("✓ Exactly 117 paired students loaded")
print("✓ 61 Not-At-Risk observations confirmed")
print("✓ 56 At-Risk observations confirmed")
print("✓ SP-XGBoost probabilities validated")
print("✓ Random Forest probabilities validated")
print("✓ Student ordering preserved")

# ======================================================================
# 3. DELONG SUPPORT FUNCTIONS
# ======================================================================

def compute_midrank(x):
    """
    Compute midranks required by the fast DeLong algorithm.

    Parameters
    ----------
    x : array-like
        Prediction scores.

    Returns
    -------
    np.ndarray
        Midrank values.
    """

    x = np.asarray(
        x,
        dtype=float
    )

    order = np.argsort(x)

    sorted_x = x[order]

    n = len(x)

    midranks = np.zeros(
        n,
        dtype=float
    )

    i = 0

    while i < n:

        j = i

        while (
            j < n and
            sorted_x[j] == sorted_x[i]
        ):
            j += 1

        midrank_value = (
            0.5 *
            (
                i +
                j -
                1
            )
        )

        midranks[
            i:j
        ] = midrank_value

        i = j

    result = np.empty(
        n,
        dtype=float
    )

    result[
        order
    ] = midranks + 1.0

    return result


def covariance_matrix(values):

    """
    Return a consistently two-dimensional covariance matrix.
    This prevents scalar covariance output from causing downstream
    indexing problems when only one classifier is present.
    """

    values = np.asarray(
        values,
        dtype=float
    )

    covariance = np.cov(
        values,
        bias=False
    )

    return np.atleast_2d(
        covariance
    )


def fast_delong(
    predictions_sorted_transposed,
    label_1_count
):
    """
    Fast DeLong covariance estimation for correlated ROC-AUCs.
    Parameters
    ----------
    predictions_sorted_transposed : np.ndarray
        Shape = (number_of_models, number_of_observations).
        Positive-class observations must appear first.

    label_1_count : int
        Number of positive observations.

    Returns
    -------
    aucs : np.ndarray
        ROC-AUC estimates.

    delong_covariance : np.ndarray
        Covariance matrix of the ROC-AUC estimates.
    """

    predictions_sorted_transposed = np.asarray(
        predictions_sorted_transposed,
        dtype=float
    )

    m = int(
        label_1_count
    )

    n = (
        predictions_sorted_transposed.shape[1] -
        m
    )

    k = (
        predictions_sorted_transposed.shape[0]
    )

    assert m > 1
    assert n > 1

    positive_predictions = (
        predictions_sorted_transposed[
            :,
            :m
        ]
    )

    negative_predictions = (
        predictions_sorted_transposed[
            :,
            m:
        ]
    )

    tx = np.empty(
        [
            k,
            m
        ],
        dtype=float
    )

    ty = np.empty(
        [
            k,
            n
        ],
        dtype=float
    )

    tz = np.empty(
        [
            k,
            m + n
        ],
        dtype=float
    )

    for model_index in range(k):

        tx[
            model_index,
            :
        ] = compute_midrank(
            positive_predictions[
                model_index,
                :
            ]
        )

        ty[
            model_index,
            :
        ] = compute_midrank(
            negative_predictions[
                model_index,
                :
            ]
        )

        tz[
            model_index,
            :
        ] = compute_midrank(
            predictions_sorted_transposed[
                model_index,
                :
            ]
        )

    aucs = (
        tz[
            :,
            :m
        ].sum(
            axis=1
        ) /
        (
            m *
            n
        )
        -
        (
            m + 1.0
        ) /
        (
            2.0 *
            n
        )
    )

    v01 = (
        tz[
            :,
            :m
        ]
        -
        tx
    ) / n

    v10 = (
        1.0
        -
        (
            tz[
                :,
                m:
            ]
            -
            ty
        ) / m
    )

    sx = covariance_matrix(
        v01
    )

    sy = covariance_matrix(
        v10
    )

    delong_covariance = (
        sx / m +
        sy / n
    )

    return (
        aucs,
        delong_covariance
    )


def paired_delong_test(
    y_true,
    model_1_scores,
    model_2_scores
):
    """
    Perform a two-sided paired DeLong test comparing two correlated
    ROC-AUC values measured on the same observations.
    """

    y_true = np.asarray(
        y_true,
        dtype=int
    )

    model_1_scores = np.asarray(
        model_1_scores,
        dtype=float
    )

    model_2_scores = np.asarray(
        model_2_scores,
        dtype=float
    )

    assert len(y_true) == len(
        model_1_scores
    )

    assert len(y_true) == len(
        model_2_scores
    )

    assert set(
        np.unique(y_true)
    ) == {0, 1}

    # --------------------------------------------------------------
    # Sort positives first, as required by fast DeLong
    # --------------------------------------------------------------

    order = np.argsort(
        -y_true
    )

    y_sorted = (
        y_true[
            order
        ]
    )

    model_1_sorted = (
        model_1_scores[
            order
        ]
    )

    model_2_sorted = (
        model_2_scores[
            order
        ]
    )

    positive_count = int(
        np.sum(
            y_sorted == 1
        )
    )

    predictions_sorted = np.vstack(
        [
            model_1_sorted,
            model_2_sorted
        ]
    )

    aucs, covariance = fast_delong(
        predictions_sorted,
        positive_count
    )

    # --------------------------------------------------------------
    # Difference in correlated ROC-AUC estimates
    # --------------------------------------------------------------

    auc_difference = (
        aucs[0] -
        aucs[1]
    )

    contrast = np.array(
        [
            1.0,
            -1.0
        ]
    )

    variance_difference = float(
        contrast
        @ covariance
        @ contrast.T
    )

    # Numerical floating-point safeguard
    if (
        variance_difference < 0 and
        np.isclose(
            variance_difference,
            0.0,
            atol=1e-15
        )
    ):
        variance_difference = 0.0

    if variance_difference < 0:

        raise ValueError(
            "Estimated DeLong variance of the AUC difference "
            "is negative."
        )

    standard_error = np.sqrt(
        variance_difference
    )

    # --------------------------------------------------------------
    # Test statistic and two-sided p-value
    # --------------------------------------------------------------

    if np.isclose(
        standard_error,
        0.0,
        atol=1e-15
    ):

        if np.isclose(
            auc_difference,
            0.0,
            atol=1e-15
        ):

            z_statistic = 0.0
            p_value = 1.0

        else:

            z_statistic = (
                np.inf
                if auc_difference > 0
                else -np.inf
            )

            p_value = 0.0

    else:

        z_statistic = (
            auc_difference /
            standard_error
        )

        p_value = (
            2.0 *
            norm.sf(
                abs(
                    z_statistic
                )
            )
        )

    return {
        "AUC_Model_1":
            float(
                aucs[0]
            ),

        "AUC_Model_2":
            float(
                aucs[1]
            ),

        "AUC_Difference":
            float(
                auc_difference
            ),

        "Variance_Difference":
            float(
                variance_difference
            ),

        "Standard_Error":
            float(
                standard_error
            ),

        "Z_Statistic":
            float(
                z_statistic
            ),

        "P_Value":
            float(
                p_value
            ),

        "Covariance_Matrix":
            covariance
    }


print()
print("✓ DeLong support functions defined")

# ======================================================================
# 4. INDEPENDENT ROC-AUC RECOMPUTATION
# ======================================================================

sp_auc_sklearn = roc_auc_score(
    y_true,
    sp_probability
)

rf_auc_sklearn = roc_auc_score(
    y_true,
    rf_probability
)

observed_difference = (
    sp_auc_sklearn -
    rf_auc_sklearn
)

print()
print("=" * 70)
print("OBSERVED MATCHED SEED-42 ROC-AUC")
print("=" * 70)

print(
    f"SP-XGBoost ROC-AUC: "
    f"{sp_auc_sklearn:.8f}"
)

print(
    f"Random Forest ROC-AUC: "
    f"{rf_auc_sklearn:.8f}"
)

print(
    f"Observed difference: "
    f"{observed_difference:+.8f}"
)

# ======================================================================
# 5. RUN PAIRED DELONG TEST
# ======================================================================

delong_result = paired_delong_test(
    y_true=y_true,
    model_1_scores=sp_probability,
    model_2_scores=rf_probability
)

sp_auc_delong = (
    delong_result[
        "AUC_Model_1"
    ]
)

rf_auc_delong = (
    delong_result[
        "AUC_Model_2"
    ]
)

auc_difference = (
    delong_result[
        "AUC_Difference"
    ]
)

variance_difference = (
    delong_result[
        "Variance_Difference"
    ]
)

standard_error = (
    delong_result[
        "Standard_Error"
    ]
)

z_statistic = (
    delong_result[
        "Z_Statistic"
    ]
)

p_value = (
    delong_result[
        "P_Value"
    ]
)

delong_covariance = (
    delong_result[
        "Covariance_Matrix"
    ]
)

# ======================================================================
# 6. CROSS-VALIDATE DELONG AUC VALUES AGAINST SKLEARN
# ======================================================================

assert np.isclose(
    sp_auc_delong,
    sp_auc_sklearn,
    atol=1e-12
)

assert np.isclose(
    rf_auc_delong,
    rf_auc_sklearn,
    atol=1e-12
)

assert np.isclose(
    auc_difference,
    observed_difference,
    atol=1e-12
)

assert delong_covariance.shape == (
    2,
    2
)

assert np.isfinite(
    delong_covariance
).all()

assert variance_difference >= 0
assert standard_error >= 0

assert (
    np.isfinite(
        p_value
    )
    and
    0.0 <= p_value <= 1.0
)

print()
print("✓ DeLong ROC-AUC values reproduce sklearn ROC-AUC values")
print("✓ Paired covariance matrix validated")
print("✓ AUC difference validated")
print("✓ DeLong variance validated")
print("✓ Two-sided p-value validated")

# ======================================================================
# 7. STATISTICAL DECISION
# ======================================================================

ALPHA = 0.05

if p_value < ALPHA:

    statistical_decision = (
        "Reject H0"
    )

    significance_status = (
        "Statistically significant difference"
    )

else:

    statistical_decision = (
        "Fail to reject H0"
    )

    significance_status = (
        "No statistically significant difference"
    )

# ======================================================================
# 8. DISPLAY DELONG RESULTS
# ======================================================================

print()
print("=" * 70)
print("PAIRED DELONG ROC-AUC TEST RESULTS")
print("=" * 70)

print(
    f"SP-XGBoost ROC-AUC:       "
    f"{sp_auc_delong:.8f}"
)

print(
    f"Random Forest ROC-AUC:    "
    f"{rf_auc_delong:.8f}"
)

print(
    f"ROC-AUC difference:       "
    f"{auc_difference:+.8f}"
)

print(
    f"Variance of difference:   "
    f"{variance_difference:.10f}"
)

print(
    f"Standard error:           "
    f"{standard_error:.8f}"
)

print(
    f"DeLong z-statistic:       "
    f"{z_statistic:.6f}"
)

print(
    f"Two-sided p-value:        "
    f"{p_value:.6f}"
)

print(
    f"Alpha:                    "
    f"{ALPHA:.2f}"
)

print(
    f"Decision:                 "
    f"{statistical_decision}"
)

print(
    f"Interpretation:           "
    f"{significance_status}"
)

# ======================================================================
# 9. DISPLAY DELONG COVARIANCE MATRIX
# ======================================================================

delong_covariance_df = pd.DataFrame(
    delong_covariance,
    index=[
        "SP_XGBoost",
        "Random_Forest"
    ],
    columns=[
        "SP_XGBoost",
        "Random_Forest"
    ]
)

print()
print("=" * 70)
print("DELONG ROC-AUC COVARIANCE MATRIX")
print("=" * 70)

print(
    delong_covariance_df.to_string(
        float_format=lambda x: f"{x:.10f}"
    )
)

# ======================================================================
# 10. BUILD SAVED STATISTICAL RESULT
# ======================================================================

delong_results_df = pd.DataFrame(
    [
        {
            "Comparison":
                "SP-XGBoost vs Random Forest",

            "Seed":
                42,

            "Test_Set_N":
                117,

            "Positive_N":
                int(
                    np.sum(
                        y_true == 1
                    )
                ),

            "Negative_N":
                int(
                    np.sum(
                        y_true == 0
                    )
                ),

            "SP_XGBoost_ROC_AUC":
                sp_auc_delong,

            "Random_Forest_ROC_AUC":
                rf_auc_delong,

            "ROC_AUC_Difference_SP_minus_RF":
                auc_difference,

            "DeLong_Variance_Difference":
                variance_difference,

            "DeLong_Standard_Error":
                standard_error,

            "DeLong_Z":
                z_statistic,

            "P_Value_Two_Sided":
                p_value,

            "Alpha":
                ALPHA,

            "Decision":
                statistical_decision,

            "Significance":
                significance_status
        }
    ]
)

# ======================================================================
# 11. SAVE DELONG RESULTS
# ======================================================================

delong_results_df.to_csv(
    DELONG_RESULTS_PATH,
    index=False
)

assert DELONG_RESULTS_PATH.exists()

reloaded_delong_results = pd.read_csv(
    DELONG_RESULTS_PATH
)

assert len(
    reloaded_delong_results
) == 1

assert np.isclose(
    reloaded_delong_results.loc[
        0,
        "SP_XGBoost_ROC_AUC"
    ],
    sp_auc_delong
)

assert np.isclose(
    reloaded_delong_results.loc[
        0,
        "Random_Forest_ROC_AUC"
    ],
    rf_auc_delong
)

print()
print("=" * 70)
print("DELONG RESULT ARTIFACT")
print("=" * 70)

print(
    f"✓ Saved: "
    f"{DELONG_RESULTS_PATH}"
)

print(
    "✓ Saved statistical result successfully reloaded"
)

# ======================================================================
# 12. INFERENTIAL SAFEGUARDS
# ======================================================================

print()
print("=" * 70)
print("INFERENTIAL SAFEGUARDS")
print("=" * 70)

print(
    "✓ Comparison is paired at the student level"
)

print(
    "✓ Same 117 independent-test students used for both models"
)

print(
    "✓ DeLong test uses continuous predicted probabilities"
)

print(
    "✓ Random seeds are NOT treated as independent samples"
)

print(
    "✓ No paired t-test performed on three seed-level ROC-AUC values"
)

print(
    "✓ Statistical inference refers specifically to the "
    "matched seed-42 comparison"
)

print(
    "✓ Failure to reject H0 will NOT be interpreted as "
    "proof that the models are equivalent"
)

print(
    "✓ Original final SP-XGBoost evaluation remains "
    "a separate evaluation"
)
print(
    "✓ Notebook 12 three-seed robustness remains "
    "a separate robustness analysis"
)

print(
    "✓ Notebook 14 matched three-seed means remain "
    "descriptive robustness evidence"
)

# ======================================================================
# 13. CELL 4 FINAL VALIDATION
# ======================================================================

print()
print("=" * 70)
print("CELL 4 FINAL VALIDATION")
print("=" * 70)

print("✓ 117 paired observations used")
print("✓ 56 positive observations used")
print("✓ 61 negative observations used")
print("✓ SP-XGBoost probability scores used")
print("✓ Random Forest probability scores used")
print("✓ Correlated ROC-AUC covariance estimated")
print("✓ Paired DeLong test completed")
print("✓ Two-sided hypothesis test completed")
print("✓ Statistical decision recorded")
print("✓ DeLong result saved")
print("✓ No model trained")
print("✓ No prediction regenerated")
print("✓ No hyperparameter optimisation performed")
print("✓ No feature selection performed")
print("✓ No dataset modified")
print("✓ Notebook 12 remains locked")
print("✓ Notebook 14 remains locked")

print()
print("=" * 70)
print("✓ CELL 4 — PAIRED DELONG TEST COMPLETED")
print("=" * 70)
```

```python
# ======================================================================
# NOTEBOOK 15 — PAIRED STATISTICAL COMPARISON
#               AND UNCERTAINTY ANALYSIS
#
# CELL 5 — PAIRED STRATIFIED BOOTSTRAP
#          95% CONFIDENCE INTERVALS
#
# PURPOSE:
# Estimate paired bootstrap uncertainty for:
#   ROC-AUC
#   PR-AUC
#   F1
#   Precision
#   Recall
#   Specificity
#   Accuracy
#
# Difference convention:
#   SP-XGBoost minus Random Forest
#
# Bootstrap unit:
#   Student
#
# NO MODEL TRAINING
# NO HYPERPARAMETER OPTIMISATION
# NO FEATURE SELECTION
# NO THRESHOLD OPTIMISATION
# ======================================================================

print("=" * 70)
print("NOTEBOOK 15 — PAIRED STRATIFIED BOOTSTRAP")
print("95% CONFIDENCE INTERVALS")
print("=" * 70)

# ======================================================================
# 1. CONFIGURATION
# ======================================================================

BOOTSTRAP_SEED = 42

N_BOOTSTRAP = 5000

CONFIDENCE_LEVEL = 0.95

ALPHA_CI = (
    1.0 -
    CONFIDENCE_LEVEL
)

LOWER_PERCENTILE = (
    100 *
    ALPHA_CI /
    2
)

UPPER_PERCENTILE = (
    100 *
    (
        1 -
        ALPHA_CI / 2
    )
)

assert BOOTSTRAP_SEED == 42
assert N_BOOTSTRAP == 5000
assert CONFIDENCE_LEVEL == 0.95

rng = np.random.default_rng(
    BOOTSTRAP_SEED
)

print()
print("=" * 70)
print("BOOTSTRAP CONFIGURATION")
print("=" * 70)

print(
    f"Bootstrap seed:       "
    f"{BOOTSTRAP_SEED}"
)

print(
    f"Bootstrap iterations: "
    f"{N_BOOTSTRAP}"
)

print(
    f"Confidence level:     "
    f"{CONFIDENCE_LEVEL:.0%}"
)

print(
    "Bootstrap design:     "
    "Paired stratified resampling"
)

print(
    "Resampling unit:      "
    "Student"
)

# ======================================================================
# 2. LOAD LOCKED PAIRED ARTIFACT
# ======================================================================

PROJECT_DIR = Path("..")

RESULTS_DIR = (
    PROJECT_DIR /
    "results"
)

NOTEBOOK15_DIR = (
    RESULTS_DIR /
    "notebook15_statistics"
)

PAIRED_SEED42_PATH = (
    NOTEBOOK15_DIR /
    "sp_xgboost_vs_random_forest_seed42_paired_predictions.csv"
)

BOOTSTRAP_SUMMARY_PATH = (
    NOTEBOOK15_DIR /
    "sp_xgboost_vs_random_forest_seed42_bootstrap_ci.csv"
)

BOOTSTRAP_DISTRIBUTION_PATH = (
    NOTEBOOK15_DIR /
    "sp_xgboost_vs_random_forest_seed42_bootstrap_distributions.csv"
)

assert PAIRED_SEED42_PATH.exists()

paired_df = pd.read_csv(
    PAIRED_SEED42_PATH
)

assert len(
    paired_df
) == 117

# ======================================================================
# 3. EXTRACT PAIRED DATA
# ======================================================================

y_true = (
    paired_df[
        "RISK_BINARY_ACTUAL"
    ]
    .astype(int)
    .to_numpy()
)

sp_prob = (
    paired_df[
        "SP_XGBoost_Probability"
    ]
    .astype(float)
    .to_numpy()
)

sp_pred = (
    paired_df[
        "SP_XGBoost_Prediction"
    ]
    .astype(int)
    .to_numpy()
)

rf_prob = (
    paired_df[
        "Random_Forest_Probability"
    ]
    .astype(float)
    .to_numpy()
)

rf_pred = (
    paired_df[
        "Random_Forest_Prediction"
    ]
    .astype(int)
    .to_numpy()
)

assert len(y_true) == 117

assert set(
    np.unique(y_true)
) == {0, 1}

assert np.sum(
    y_true == 1
) == 56

assert np.sum(
    y_true == 0
) == 61

print()
print("=" * 70)
print("PAIRED DATA VALIDATION")
print("=" * 70)

print("✓ 117 paired students loaded")
print("✓ 56 At-Risk students confirmed")
print("✓ 61 Not-at-Risk students confirmed")
print("✓ Both model probabilities aligned")
print("✓ Both model predictions aligned")

# ======================================================================
# 4. METRIC FUNCTION
# ======================================================================

def calculate_binary_metrics(
    y,
    probability,
    prediction
):
    """
    Calculate all seven binary performance metrics.
    """

    tn, fp, fn, tp = (
        confusion_matrix(
            y,
            prediction,
            labels=[0, 1]
        )
        .ravel()
    )

    specificity = (
        tn / (tn + fp)
        if (tn + fp) > 0
        else np.nan
    )

    return {
        "ROC_AUC":
            roc_auc_score(
                y,
                probability
            ),

        "PR_AUC":
            average_precision_score(
                y,
                probability
            ),

        "F1":
            f1_score(
                y,
                prediction,
                zero_division=0
            ),

        "Precision":
            precision_score(
                y,
                prediction,
                zero_division=0
            ),

        "Recall":
            recall_score(
                y,
                prediction,
                zero_division=0
            ),

        "Specificity":
            specificity,

        "Accuracy":
            accuracy_score(
                y,
                prediction
            )
    }


METRICS = [
    "ROC_AUC",
    "PR_AUC",
    "F1",
    "Precision",
    "Recall",
    "Specificity",
    "Accuracy"
]

print()
print("✓ Seven-metric evaluation function defined")

# ======================================================================
# 5. OBSERVED PAIRED PERFORMANCE
# ======================================================================

sp_observed = calculate_binary_metrics(
    y_true,
    sp_prob,
    sp_pred
)

rf_observed = calculate_binary_metrics(
    y_true,
    rf_prob,
    rf_pred
)

observed_differences = {
    metric:
        (
            sp_observed[metric]
            -
            rf_observed[metric]
        )
    for metric in METRICS
}

print()
print("=" * 70)
print("OBSERVED SEED-42 DIFFERENCES")
print("=" * 70)

for metric in METRICS:

    print(
        f"{metric:<15}: "
        f"{observed_differences[metric]:+.6f}"
    )

# ======================================================================
# 6. STRATIFIED STUDENT INDICES
# ======================================================================

positive_indices = np.where(
    y_true == 1
)[0]

negative_indices = np.where(
    y_true == 0
)[0]

N_POSITIVE = len(
    positive_indices
)

N_NEGATIVE = len(
    negative_indices
)

assert N_POSITIVE == 56
assert N_NEGATIVE == 61

print()
print("=" * 70)
print("STRATIFIED BOOTSTRAP STRUCTURE")
print("=" * 70)

print(
    f"Positive stratum: "
    f"{N_POSITIVE}"
)

print(
    f"Negative stratum: "
    f"{N_NEGATIVE}"
)

print(
    "✓ Each bootstrap sample preserves "
    "the original class counts"
)

print(
    "✓ Model outputs remain paired "
    "student-by-student"
)

# ======================================================================
# 7. BOOTSTRAP RESULT CONTAINERS
# ======================================================================

bootstrap_differences = {
    metric:
        np.empty(
            N_BOOTSTRAP,
            dtype=float
        )
    for metric in METRICS
}

# ======================================================================
# 8. RUN PAIRED STRATIFIED BOOTSTRAP
# ======================================================================

print()
print("=" * 70)
print("RUNNING PAIRED STRATIFIED BOOTSTRAP")
print("=" * 70)

for bootstrap_index in range(
    N_BOOTSTRAP
):

    # --------------------------------------------------------------
    # Resample positive students with replacement
    # --------------------------------------------------------------

    sampled_positive = rng.choice(
        positive_indices,
        size=N_POSITIVE,
        replace=True
    )

    # --------------------------------------------------------------
    # Resample negative students with replacement
    # --------------------------------------------------------------

    sampled_negative = rng.choice(
        negative_indices,
        size=N_NEGATIVE,
        replace=True
    )

    # --------------------------------------------------------------
    # Combine strata
    # --------------------------------------------------------------

    sampled_indices = np.concatenate(
        [
            sampled_positive,
            sampled_negative
        ]
    )

    # --------------------------------------------------------------
    # Paired bootstrap sample
    # --------------------------------------------------------------

    y_boot = (
        y_true[
            sampled_indices
        ]
    )

    sp_prob_boot = (
        sp_prob[
            sampled_indices
        ]
    )

    sp_pred_boot = (
        sp_pred[
            sampled_indices
        ]
    )

    rf_prob_boot = (
        rf_prob[
            sampled_indices
        ]
    )

    rf_pred_boot = (
        rf_pred[
            sampled_indices
        ]
    )

    # --------------------------------------------------------------
    # Calculate metrics for both models
    # --------------------------------------------------------------

    sp_boot_metrics = calculate_binary_metrics(
        y_boot,
        sp_prob_boot,
        sp_pred_boot
    )

    rf_boot_metrics = calculate_binary_metrics(
        y_boot,
        rf_prob_boot,
        rf_pred_boot
    )

    # --------------------------------------------------------------
    # Store paired differences
    # --------------------------------------------------------------

    for metric in METRICS:

        bootstrap_differences[
            metric
        ][
            bootstrap_index
        ] = (
            sp_boot_metrics[
                metric
            ]
            -
            rf_boot_metrics[
                metric
            ]
        )

print()
print(
    f"✓ {N_BOOTSTRAP} paired bootstrap "
    "replicates completed"
)

# ======================================================================
# 9. BOOTSTRAP VALIDATION
# ======================================================================

for metric in METRICS:

    assert np.isfinite(
        bootstrap_differences[
            metric
        ]
    ).all()

    assert len(
        bootstrap_differences[
            metric
        ]
    ) == N_BOOTSTRAP

print()
print("✓ All bootstrap metric differences are finite")
print("✓ All seven metrics contain 5000 paired replicates")

# ======================================================================
# 10. BUILD BOOTSTRAP CONFIDENCE-INTERVAL SUMMARY
# ======================================================================

bootstrap_summary_rows = []

for metric in METRICS:

    distribution = (
        bootstrap_differences[
            metric
        ]
    )

    bootstrap_mean = (
        np.mean(
            distribution
        )
    )

    bootstrap_sd = (
        np.std(
            distribution,
            ddof=1
        )
    )

    ci_lower = (
        np.percentile(
            distribution,
            LOWER_PERCENTILE
        )
    )

    ci_upper = (
        np.percentile(
            distribution,
            UPPER_PERCENTILE
        )
    )

    # --------------------------------------------------------------
    # CI relationship with zero
    # --------------------------------------------------------------

    if (
        ci_lower > 0
    ):

        ci_zero_status = (
            "Entirely above zero"
        )

    elif (
        ci_upper < 0
    ):

        ci_zero_status = (
            "Entirely below zero"
        )

    else:

        ci_zero_status = (
            "Includes zero"
        )

    bootstrap_summary_rows.append({
        "Metric":
            metric,

        "SP_XGBoost_Observed":
            sp_observed[
                metric
            ],

        "Random_Forest_Observed":
            rf_observed[
                metric
            ],

        "Observed_Difference_SP_minus_RF":
            observed_differences[
                metric
            ],

        "Bootstrap_Mean_Difference":
            bootstrap_mean,

        "Bootstrap_SE":
            bootstrap_sd,

        "CI_95_Lower":
            ci_lower,

        "CI_95_Upper":
            ci_upper,

        "CI_Zero_Status":
            ci_zero_status,

        "Bootstrap_Iterations":
            N_BOOTSTRAP
    })

bootstrap_summary_df = pd.DataFrame(
    bootstrap_summary_rows
)

assert len(
    bootstrap_summary_df
) == 7

# ======================================================================
# 11. DISPLAY BOOTSTRAP RESULTS
# ======================================================================

print()
print("=" * 70)
print("PAIRED BOOTSTRAP 95% CONFIDENCE INTERVALS")
print("=" * 70)

print(
    bootstrap_summary_df[
        [
            "Metric",
            "SP_XGBoost_Observed",
            "Random_Forest_Observed",
            "Observed_Difference_SP_minus_RF",
            "Bootstrap_SE",
            "CI_95_Lower",
            "CI_95_Upper",
            "CI_Zero_Status"
        ]
    ].to_string(
        index=False,
        formatters={
            "SP_XGBoost_Observed":
                "{:.6f}".format,

            "Random_Forest_Observed":
                "{:.6f}".format,

            "Observed_Difference_SP_minus_RF":
                "{:+.6f}".format,

            "Bootstrap_SE":
                "{:.6f}".format,

            "CI_95_Lower":
                "{:+.6f}".format,

            "CI_95_Upper":
                "{:+.6f}".format
        }
    )
)

# ======================================================================
# 12. PRIMARY ROC-AUC BOOTSTRAP RESULT
# ======================================================================

roc_bootstrap_row = (
    bootstrap_summary_df.loc[
        bootstrap_summary_df[
            "Metric"
        ] == "ROC_AUC"
    ]
    .iloc[0]
)

print()
print("=" * 70)
print("PRIMARY ROC-AUC BOOTSTRAP INTERVAL")
print("=" * 70)

print(
    f"Observed ΔROC-AUC: "
    f"{roc_bootstrap_row['Observed_Difference_SP_minus_RF']:+.8f}"
)

print(
    f"Bootstrap SE:       "
    f"{roc_bootstrap_row['Bootstrap_SE']:.8f}"
)

print(
    f"95% CI:             "
    f"["
    f"{roc_bootstrap_row['CI_95_Lower']:+.8f}, "
    f"{roc_bootstrap_row['CI_95_Upper']:+.8f}"
    f"]"
)

print(
    f"Zero status:        "
    f"{roc_bootstrap_row['CI_Zero_Status']}"
)

# ======================================================================
# 13. COMPARE WITH DELONG RESULT
# ======================================================================

DELONG_RESULTS_PATH = (
    NOTEBOOK15_DIR /
    "sp_xgboost_vs_random_forest_seed42_delong_test.csv"
)

assert DELONG_RESULTS_PATH.exists()

delong_df = pd.read_csv(
    DELONG_RESULTS_PATH
)

assert len(
    delong_df
) == 1

delong_difference = float(
    delong_df.loc[
        0,
        "ROC_AUC_Difference_SP_minus_RF"
    ]
)

delong_p_value = float(
    delong_df.loc[
        0,
        "P_Value_Two_Sided"
    ]
)

assert np.isclose(
    delong_difference,
    observed_differences[
        "ROC_AUC"
    ],
    atol=1e-12
)

print()
print("=" * 70)
print("DELONG / BOOTSTRAP CONSISTENCY")
print("=" * 70)

print(
    f"DeLong observed ΔROC-AUC: "
    f"{delong_difference:+.8f}"
)

print(
    f"DeLong p-value:           "
    f"{delong_p_value:.6f}"
)

print(
    f"Bootstrap 95% CI:         "
    f"["
    f"{roc_bootstrap_row['CI_95_Lower']:+.8f}, "
    f"{roc_bootstrap_row['CI_95_Upper']:+.8f}"
    f"]"
)

if (
    roc_bootstrap_row[
        "CI_95_Lower"
    ] <= 0 <=
    roc_bootstrap_row[
        "CI_95_Upper"
    ]
):

    print(
        "✓ Bootstrap interval includes zero"
    )

else:

    print(
        "⚠ Bootstrap interval excludes zero"
    )

# ======================================================================
# 14. SAVE FULL BOOTSTRAP DISTRIBUTIONS
# ======================================================================

bootstrap_distribution_df = pd.DataFrame({
    f"{metric}_Difference":
        bootstrap_differences[
            metric
        ]
    for metric in METRICS
})

bootstrap_distribution_df.insert(
    0,
    "Bootstrap_Iteration",
    np.arange(
        1,
        N_BOOTSTRAP + 1
    )
)

assert len(
    bootstrap_distribution_df
) == N_BOOTSTRAP

bootstrap_distribution_df.to_csv(
    BOOTSTRAP_DISTRIBUTION_PATH,
    index=False
)

# ======================================================================
# 15. SAVE BOOTSTRAP SUMMARY
# ======================================================================

bootstrap_summary_df.to_csv(
    BOOTSTRAP_SUMMARY_PATH,
    index=False
)

assert BOOTSTRAP_SUMMARY_PATH.exists()

assert BOOTSTRAP_DISTRIBUTION_PATH.exists()

# ----------------------------------------------------------------------
# Reload validation
# ----------------------------------------------------------------------

bootstrap_summary_reload = pd.read_csv(
    BOOTSTRAP_SUMMARY_PATH
)

bootstrap_distribution_reload = pd.read_csv(
    BOOTSTRAP_DISTRIBUTION_PATH
)

assert len(
    bootstrap_summary_reload
) == 7

assert len(
    bootstrap_distribution_reload
) == N_BOOTSTRAP

print()
print("=" * 70)
print("BOOTSTRAP ARTIFACTS SAVED")
print("=" * 70)

print(
    f"✓ Summary:\n"
    f"  {BOOTSTRAP_SUMMARY_PATH}"
)

print()

print(
    f"✓ Full distributions:\n"
    f"  {BOOTSTRAP_DISTRIBUTION_PATH}"
)

print()

print(
    "✓ Saved bootstrap artifacts "
    "successfully reloaded"
)

# ======================================================================
# 16. INTERPRETIVE SAFEGUARDS
# ======================================================================

print()
print("=" * 70)
print("BOOTSTRAP INTERPRETIVE SAFEGUARDS")
print("=" * 70)

print(
    "✓ Bootstrap resampling was paired at the student level"
)

print(
    "✓ Class counts were preserved through stratified resampling"
)

print(
    "✓ SP-XGBoost and RF outputs were never resampled independently"
)

print(
    "✓ The DeLong test remains the primary inferential "
    "test for ROC-AUC"
)

print(
    "✓ Bootstrap intervals quantify uncertainty "
    "around observed metric differences"
)

print(
    "✓ An interval containing zero indicates that "
    "the corresponding difference is compatible with zero"
)

print(
    "✓ An interval excluding zero is descriptive evidence "
    "of a consistently signed difference under this "
    "bootstrap procedure"
)

print(
    "✓ No equivalence claim will be made merely because "
    "a confidence interval includes zero"
)

print(
    "✓ No seed-level pseudo-replication used"
)

# ======================================================================
# 17. CELL 5 FINAL VALIDATION
# ======================================================================

print()
print("=" * 70)
print("CELL 5 FINAL VALIDATION")
print("=" * 70)

print("✓ 5000 paired bootstrap replicates completed")
print("✓ 117 students used as resampling units")
print("✓ Stratified class structure preserved")
print("✓ ROC-AUC uncertainty quantified")
print("✓ PR-AUC uncertainty quantified")
print("✓ F1 uncertainty quantified")
print("✓ Precision uncertainty quantified")
print("✓ Recall uncertainty quantified")
print("✓ Specificity uncertainty quantified")
print("✓ Accuracy uncertainty quantified")
print("✓ 95% percentile confidence intervals calculated")
print("✓ DeLong/ROC-AUC consistency checked")
print("✓ Bootstrap summary saved")
print("✓ Full bootstrap distributions saved")
print("✓ No model trained")
print("✓ No prediction regenerated")
print("✓ No hyperparameter optimisation performed")
print("✓ No feature selection performed")
print("✓ No threshold optimisation performed")
print("✓ No dataset modified")
print("✓ Notebook 12 remains locked")
print("✓ Notebook 14 remains locked")

print()
print("=" * 70)
print("✓ CELL 5 — PAIRED BOOTSTRAP ANALYSIS COMPLETED")
print("=" * 70)
```

```python
# ======================================================================
# NOTEBOOK 15 — PAIRED STATISTICAL COMPARISON
#               AND UNCERTAINTY ANALYSIS
#
# CELL 6 — EXACT MCNEMAR PAIRED CLASSIFICATION-ERROR TEST
#
# PURPOSE:
# Compare SP-XGBoost and Random Forest classification errors
# on the SAME 117 independent-test students.
#
# PRIMARY QUESTION:
# Do the two models have different probabilities of making
# classification errors on the same students?
#
# TEST:
# Exact two-sided McNemar test
#
# PAIRING UNIT:
# Individual student
#
# NO MODEL TRAINING
# NO HYPERPARAMETER OPTIMISATION
# NO FEATURE SELECTION
# NO THRESHOLD OPTIMISATION
# ======================================================================

print("=" * 70)
print("NOTEBOOK 15 — EXACT MCNEMAR")
print("PAIRED CLASSIFICATION-ERROR COMPARISON")
print("=" * 70)

# ----------------------------------------------------------------------
# REQUIRED IMPORTS
# ----------------------------------------------------------------------

from pathlib import Path

import numpy as np
import pandas as pd

from statsmodels.stats.contingency_tables import mcnemar

print()
print("✓ Required McNemar libraries imported")

# ======================================================================
# 1. CONFIGURATION
# ======================================================================

ALPHA = 0.05

assert ALPHA == 0.05

print()
print("=" * 70)
print("STATISTICAL CONFIGURATION")
print("=" * 70)

print(f"Alpha:             {ALPHA}")
print("Test:              Exact McNemar")
print("Alternative:       Two-sided")
print("Pairing unit:      Student")
print("Comparison:        SP-XGBoost vs Random Forest")
print("Evaluation seed:   42")

# ======================================================================
# 2. LOAD LOCKED PAIRED PREDICTION ARTIFACT
# ======================================================================

PROJECT_DIR = Path("..")

RESULTS_DIR = (
    PROJECT_DIR /
    "results"
)

NOTEBOOK15_DIR = (
    RESULTS_DIR /
    "notebook15_statistics"
)

PAIRED_PREDICTIONS_PATH = (
    NOTEBOOK15_DIR /
    "sp_xgboost_vs_random_forest_seed42_paired_predictions.csv"
)

MCNEMAR_RESULT_PATH = (
    NOTEBOOK15_DIR /
    "sp_xgboost_vs_random_forest_seed42_mcnemar_test.csv"
)

MCNEMAR_TABLE_PATH = (
    NOTEBOOK15_DIR /
    "sp_xgboost_vs_random_forest_seed42_mcnemar_table.csv"
)

assert PAIRED_PREDICTIONS_PATH.exists()

paired_df = pd.read_csv(
    PAIRED_PREDICTIONS_PATH
)

assert len(paired_df) == 117

print()
print("=" * 70)
print("PAIRED ARTIFACT VALIDATION")
print("=" * 70)

print(
    f"Paired prediction artifact:\n"
    f"{PAIRED_PREDICTIONS_PATH}"
)

print()
print("✓ Exactly 117 paired students loaded")

# ======================================================================
# 3. EXTRACT ACTUAL AND PREDICTED LABELS
# ======================================================================

y_true = (
    paired_df[
        "RISK_BINARY_ACTUAL"
    ]
    .astype(int)
    .to_numpy()
)

sp_pred = (
    paired_df[
        "SP_XGBoost_Prediction"
    ]
    .astype(int)
    .to_numpy()
)

rf_pred = (
    paired_df[
        "Random_Forest_Prediction"
    ]
    .astype(int)
    .to_numpy()
)

assert len(y_true) == 117
assert len(sp_pred) == 117
assert len(rf_pred) == 117

assert set(
    np.unique(y_true)
) == {0, 1}

assert set(
    np.unique(sp_pred)
).issubset({0, 1})

assert set(
    np.unique(rf_pred)
).issubset({0, 1})

assert np.sum(y_true == 0) == 61
assert np.sum(y_true == 1) == 56

print("✓ Actual labels validated")
print("✓ SP-XGBoost binary predictions validated")
print("✓ Random Forest binary predictions validated")
print("✓ Student-level ordering preserved")

# ======================================================================
# 4. DETERMINE CORRECT / INCORRECT CLASSIFICATION
# ======================================================================

sp_correct = (
    sp_pred == y_true
)

rf_correct = (
    rf_pred == y_true
)

assert len(sp_correct) == 117
assert len(rf_correct) == 117

# ----------------------------------------------------------------------
# Four paired outcome groups
# ----------------------------------------------------------------------

both_correct = int(
    np.sum(
        sp_correct &
        rf_correct
    )
)

sp_correct_rf_wrong = int(
    np.sum(
        sp_correct &
        ~rf_correct
    )
)

sp_wrong_rf_correct = int(
    np.sum(
        ~sp_correct &
        rf_correct
    )
)

both_wrong = int(
    np.sum(
        ~sp_correct &
        ~rf_correct
    )
)

assert (
    both_correct
    + sp_correct_rf_wrong
    + sp_wrong_rf_correct
    + both_wrong
) == 117

discordant_total = (
    sp_correct_rf_wrong
    +
    sp_wrong_rf_correct
)

print()
print("=" * 70)
print("PAIRED CORRECTNESS TABLE")
print("=" * 70)

print()
print(
    "                         "
    "Random Forest"
)

print(
    "                     "
    "Correct    Wrong"
)

print(
    f"SP-XGBoost Correct   "
    f"{both_correct:>7}  "
    f"{sp_correct_rf_wrong:>7}"
)

print(
    f"SP-XGBoost Wrong     "
    f"{sp_wrong_rf_correct:>7}  "
    f"{both_wrong:>7}"
)

print()
print(
    f"Both models correct:                  "
    f"{both_correct}"
)

print(
    f"SP correct / RF wrong:                "
    f"{sp_correct_rf_wrong}"
)

print(
    f"SP wrong / RF correct:                "
    f"{sp_wrong_rf_correct}"
)

print(
    f"Both models wrong:                    "
    f"{both_wrong}"
)

print(
    f"Total discordant classifications:     "
    f"{discordant_total}"
)

# ======================================================================
# 5. VERIFY MODEL ACCURACIES FROM PAIRED TABLE
# ======================================================================

sp_correct_total = (
    both_correct
    +
    sp_correct_rf_wrong
)

rf_correct_total = (
    both_correct
    +
    sp_wrong_rf_correct
)

sp_accuracy = (
    sp_correct_total /
    len(y_true)
)

rf_accuracy = (
    rf_correct_total /
    len(y_true)
)

accuracy_difference = (
    sp_accuracy -
    rf_accuracy
)

assert np.isclose(
    sp_accuracy,
    0.7435897435897436,
    atol=1e-10
)

assert np.isclose(
    rf_accuracy,
    0.7264957264957265,
    atol=1e-10
)

print()
print("=" * 70)
print("PAIRED ACCURACY VALIDATION")
print("=" * 70)

print(
    f"SP-XGBoost accuracy:    "
    f"{sp_accuracy:.6f}"
)

print(
    f"Random Forest accuracy: "
    f"{rf_accuracy:.6f}"
)

print(
    f"Accuracy difference:    "
    f"{accuracy_difference:+.6f}"
)

print()
print("✓ Locked seed-42 accuracies reproduced")

# ======================================================================
# 6. MCNEMAR CONTINGENCY TABLE
# ======================================================================

# Rows:
#   SP-XGBoost correct / wrong
#
# Columns:
#   Random Forest correct / wrong
#
#              RF correct    RF wrong
# SP correct       a             b
# SP wrong         c             d

mcnemar_table = np.array(
    [
        [
            both_correct,
            sp_correct_rf_wrong
        ],
        [
            sp_wrong_rf_correct,
            both_wrong
        ]
    ],
    dtype=int
)

assert mcnemar_table.shape == (2, 2)

assert int(
    mcnemar_table.sum()
) == 117

# ======================================================================
# 7. EXACT TWO-SIDED MCNEMAR TEST
# ======================================================================

mcnemar_result = mcnemar(
    mcnemar_table,
    exact=True
)

mcnemar_statistic = float(
    mcnemar_result.statistic
)

mcnemar_p_value = float(
    mcnemar_result.pvalue
)

assert np.isfinite(
    mcnemar_statistic
)

assert np.isfinite(
    mcnemar_p_value
)

assert (
    0.0 <=
    mcnemar_p_value <=
    1.0
)

# ======================================================================
# 8. HYPOTHESIS DECISION
# ======================================================================

if mcnemar_p_value < ALPHA:

    decision = (
        "Reject H0"
    )

    interpretation = (
        "Statistically significant difference "
        "in paired classification-error rates"
    )

else:

    decision = (
        "Fail to reject H0"
    )

    interpretation = (
        "No statistically significant difference "
        "in paired classification-error rates"
    )

# ======================================================================
# 9. DIRECTION OF DISCORDANT CASES
# ======================================================================

if (
    sp_correct_rf_wrong
    >
    sp_wrong_rf_correct
):

    discordant_direction = (
        "More discordant students were correctly "
        "classified by SP-XGBoost"
    )

elif (
    sp_correct_rf_wrong
    <
    sp_wrong_rf_correct
):

    discordant_direction = (
        "More discordant students were correctly "
        "classified by Random Forest"
    )

else:

    discordant_direction = (
        "Discordant classifications were evenly split"
    )

# ======================================================================
# 10. DISPLAY EXACT MCNEMAR RESULT
# ======================================================================

print()
print("=" * 70)
print("EXACT MCNEMAR TEST RESULTS")
print("=" * 70)

print(
    f"SP correct / RF wrong:   "
    f"{sp_correct_rf_wrong}"
)

print(
    f"SP wrong / RF correct:   "
    f"{sp_wrong_rf_correct}"
)

print(
    f"Discordant total:        "
    f"{discordant_total}"
)

print(
    f"McNemar statistic:       "
    f"{mcnemar_statistic:.6f}"
)

print(
    f"Exact two-sided p-value: "
    f"{mcnemar_p_value:.6f}"
)

print(
    f"Alpha:                   "
    f"{ALPHA:.2f}"
)

print(
    f"Decision:                "
    f"{decision}"
)

print(
    f"Interpretation:          "
    f"{interpretation}"
)

print(
    f"Direction:               "
    f"{discordant_direction}"
)

# ======================================================================
# 11. SAVE MCNEMAR CONTINGENCY TABLE
# ======================================================================

mcnemar_table_df = pd.DataFrame({
    "SP_XGBoost_Status": [
        "Correct",
        "Correct",
        "Wrong",
        "Wrong"
    ],

    "Random_Forest_Status": [
        "Correct",
        "Wrong",
        "Correct",
        "Wrong"
    ],

    "Count": [
        both_correct,
        sp_correct_rf_wrong,
        sp_wrong_rf_correct,
        both_wrong
    ]
})

assert (
    mcnemar_table_df[
        "Count"
    ].sum()
) == 117

mcnemar_table_df.to_csv(
    MCNEMAR_TABLE_PATH,
    index=False
)

# ======================================================================
# 12. SAVE MCNEMAR RESULT
# ======================================================================

mcnemar_result_df = pd.DataFrame({
    "Comparison": [
        "SP-XGBoost vs Random Forest"
    ],

    "Seed": [
        42
    ],

    "N_Paired_Students": [
        117
    ],

    "Both_Correct": [
        both_correct
    ],

    "SP_Correct_RF_Wrong": [
        sp_correct_rf_wrong
    ],

    "SP_Wrong_RF_Correct": [
        sp_wrong_rf_correct
    ],

    "Both_Wrong": [
        both_wrong
    ],

    "Discordant_Total": [
        discordant_total
    ],

    "SP_XGBoost_Accuracy": [
        sp_accuracy
    ],

    "Random_Forest_Accuracy": [
        rf_accuracy
    ],

    "Accuracy_Difference_SP_minus_RF": [
        accuracy_difference
    ],

    "McNemar_Statistic": [
        mcnemar_statistic
    ],

    "P_Value_Exact_Two_Sided": [
        mcnemar_p_value
    ],

    "Alpha": [
        ALPHA
    ],

    "Decision": [
        decision
    ],

    "Interpretation": [
        interpretation
    ],

    "Discordant_Direction": [
        discordant_direction
    ]
})

mcnemar_result_df.to_csv(
    MCNEMAR_RESULT_PATH,
    index=False
)

# ======================================================================
# 13. RELOAD ARTIFACT VALIDATION
# ======================================================================

assert MCNEMAR_RESULT_PATH.exists()
assert MCNEMAR_TABLE_PATH.exists()

reloaded_mcnemar_result = pd.read_csv(
    MCNEMAR_RESULT_PATH
)

reloaded_mcnemar_table = pd.read_csv(
    MCNEMAR_TABLE_PATH
)

assert len(
    reloaded_mcnemar_result
) == 1

assert len(
    reloaded_mcnemar_table
) == 4

assert int(
    reloaded_mcnemar_table[
        "Count"
    ].sum()
) == 117

print()
print("=" * 70)
print("MCNEMAR ARTIFACTS SAVED")
print("=" * 70)

print(
    f"✓ Statistical result:\n"
    f"  {MCNEMAR_RESULT_PATH}"
)

print()

print(
    f"✓ Paired contingency table:\n"
    f"  {MCNEMAR_TABLE_PATH}"
)

print()

print(
    "✓ Saved McNemar artifacts "
    "successfully reloaded"
)

# ======================================================================
# 14. RELATIONSHIP WITH PREVIOUS NOTEBOOK 15 RESULTS
# ======================================================================

print()
print("=" * 70)
print("RELATIONSHIP WITH DELONG AND BOOTSTRAP RESULTS")
print("=" * 70)

print(
    "DeLong test:"
)

print(
    "  Tests whether paired ROC-AUC values differ"
)

print()

print(
    "Paired bootstrap:"
)

print(
    "  Quantifies uncertainty around metric differences"
)

print()

print(
    "McNemar test:"
)

print(
    "  Tests whether paired binary classification-error "
    "rates differ"
)

print()

print(
    "✓ These analyses answer related but different "
    "statistical questions"
)

# ======================================================================
# 15. STATISTICAL SAFEGUARDS
# ======================================================================

print()
print("=" * 70)
print("MCNEMAR INFERENTIAL SAFEGUARDS")
print("=" * 70)

print(
    "✓ McNemar analysis uses the same 117 students "
    "for both models"
)

print(
    "✓ Binary predictions are paired student-by-student"
)

print(
    "✓ Only discordant prediction outcomes determine "
    "the McNemar comparison"
)

print(
    "✓ Exact two-sided inference used"
)

print(
    "✓ No random seed is treated as an independent "
    "student sample"
)

print(
    "✓ Failure to reject H0 will not be interpreted "
    "as proof of model equivalence"
)

print(
    "✓ McNemar results concern overall classification "
    "errors, not ROC-AUC"
)

print(
    "✓ Recall and specificity differences remain "
    "separate metric-specific findings"
)

print(
    "✓ Original final SP-XGBoost performance remains "
    "a separate evaluation"
)

# ======================================================================
# 16. CELL 5 FINAL VALIDATION
# ======================================================================

print()
print("=" * 70)
print("CELL 5 FINAL VALIDATION")
print("=" * 70)

print("✓ 117 paired students used")
print("✓ SP-XGBoost correctness calculated")
print("✓ Random Forest correctness calculated")
print("✓ Concordant classifications calculated")
print("✓ Discordant classifications calculated")
print("✓ Locked seed-42 accuracies reproduced")
print("✓ Exact two-sided McNemar test completed")
print("✓ Statistical decision recorded")
print("✓ Discordant-case direction recorded")
print("✓ McNemar contingency table saved")
print("✓ McNemar statistical result saved")
print("✓ No model trained")
print("✓ No prediction regenerated")
print("✓ No hyperparameter optimisation performed")
print("✓ No feature selection performed")
print("✓ No threshold optimisation performed")
print("✓ No dataset modified")
print("✓ Notebook 12 remains locked")
print("✓ Notebook 14 remains locked")

print()
print("=" * 70)
print("✓ CELL 6 — EXACT MCNEMAR TEST COMPLETED")
print("=" * 70)
```
