```python
# ======================================================================
# CELL 5-NOTEBOOK 13 — FINAL MODEL COMPARISON
#              AND STATISTICAL REPORTING
#
# CELL 5 — THREE-SEED PERFORMANCE AND 15-FOLD CV SUMMARY
# ======================================================================

print("=" * 70)
print("NOTEBOOK 13 — THREE-SEED PERFORMANCE SUMMARY")
print("AND 15-FOLD CROSS-VALIDATION STABILITY")
print("=" * 70)

# ----------------------------------------------------------------------
# REQUIRED METRICS
# ----------------------------------------------------------------------

PERFORMANCE_METRICS = [
    "ROC_AUC",
    "PR_AUC",
    "F1",
    "Precision",
    "Recall",
    "Specificity",
    "Accuracy"
]

# ----------------------------------------------------------------------
# VALIDATE LOCKED INPUTS
# ----------------------------------------------------------------------

assert len(seed_test_results_df) == 3
assert len(cv_fold_results_df) == 15

assert set(
    seed_test_results_df["Seed"]
) == {42, 123, 7}

assert set(
    cv_fold_results_df["Seed"]
) == {42, 123, 7}

assert (
    cv_fold_results_df
    .groupby("Seed")
    .size()
    .eq(5)
    .all()
)

assert all(
    metric in seed_test_results_df.columns
    for metric in PERFORMANCE_METRICS
)

# ----------------------------------------------------------------------
# THREE-SEED INDEPENDENT-TEST RESULTS
# ----------------------------------------------------------------------

print()
print("=" * 70)
print("THREE-SEED INDEPENDENT-TEST PERFORMANCE")
print("=" * 70)

display_columns = [
    "Seed",
    "ROC_AUC",
    "PR_AUC",
    "F1",
    "Precision",
    "Recall",
    "Specificity",
    "Accuracy"
]

three_seed_test_table = (
    seed_test_results_df[
        display_columns
    ]
    .sort_values("Seed")
    .reset_index(drop=True)
)

print(
    three_seed_test_table.to_string(
        index=False,
        formatters={
            "ROC_AUC": "{:.4f}".format,
            "PR_AUC": "{:.4f}".format,
            "F1": "{:.4f}".format,
            "Precision": "{:.4f}".format,
            "Recall": "{:.4f}".format,
            "Specificity": "{:.4f}".format,
            "Accuracy": "{:.4f}".format
        }
    )
)

# ----------------------------------------------------------------------
# THREE-SEED MEAN AND SAMPLE SD
# ----------------------------------------------------------------------

three_seed_mean_sd = []

for metric in PERFORMANCE_METRICS:

    values = (
        seed_test_results_df[metric]
        .astype(float)
    )

    three_seed_mean_sd.append({
        "Metric": metric,
        "Mean": values.mean(),
        "SD": values.std(ddof=1),
        "Minimum": values.min(),
        "Maximum": values.max(),
        "Range": values.max() - values.min(),
        "CV_Percent": (
            values.std(ddof=1) /
            values.mean()
        ) * 100
    })

three_seed_mean_sd_df = pd.DataFrame(
    three_seed_mean_sd
)

# ----------------------------------------------------------------------
# VALIDATION
# ----------------------------------------------------------------------

assert len(three_seed_mean_sd_df) == 7

assert np.isfinite(
    three_seed_mean_sd_df[
        [
            "Mean",
            "SD",
            "Minimum",
            "Maximum",
            "Range",
            "CV_Percent"
        ]
    ].to_numpy()
).all()

# ----------------------------------------------------------------------
# DISPLAY MEAN ± SD
# ----------------------------------------------------------------------

print()
print("=" * 70)
print("THREE-SEED INDEPENDENT-TEST MEAN ± SD")
print("=" * 70)

for _, row in three_seed_mean_sd_df.iterrows():

    print(
        f"{row['Metric']:<15}: "
        f"{row['Mean']:.4f} ± {row['SD']:.4f}"
    )

# ----------------------------------------------------------------------
# MAXIMUM AND MINIMUM OBSERVED VALUES BY RANDOM SEED
# ----------------------------------------------------------------------
print()
print("=" * 70)
print("MAXIMUM AND MINIMUM OBSERVED VALUES BY RANDOM SEED")
print("=" * 70)

seed_extremes = []

for metric in PERFORMANCE_METRICS:

    # Work directly from the complete three-seed test-results DataFrame
    max_idx = seed_test_results_df[metric].idxmax()
    min_idx = seed_test_results_df[metric].idxmin()

    max_seed = int(
        seed_test_results_df.loc[max_idx, "Seed"]
    )

    min_seed = int(
        seed_test_results_df.loc[min_idx, "Seed"]
    )

    max_value = float(
        seed_test_results_df.loc[max_idx, metric]
    )

    min_value = float(
        seed_test_results_df.loc[min_idx, metric]
    )

    seed_extremes.append({
        "Metric": metric,
        "Max_Seed": max_seed,
        "Max_Value": max_value,
        "Min_Seed": min_seed,
        "Min_Value": min_value
    })

seed_extremes_df = pd.DataFrame(seed_extremes)

# Validation
assert len(seed_extremes_df) == len(PERFORMANCE_METRICS)
assert set(seed_extremes_df["Metric"]) == set(PERFORMANCE_METRICS)

print(
    seed_extremes_df.to_string(
        index=False,
        formatters={
            "Max_Value": "{:.4f}".format,
            "Min_Value": "{:.4f}".format
        }
    )
)
# ======================================================================
# 15-FOLD CV — SEED-LEVEL SUMMARY
# ======================================================================

print()
print("=" * 70)
print("5-FOLD CV STABILITY BY RANDOM SEED")
print("=" * 70)

cv_seed_summary_table = (
    cv_seed_summary_df[
        [
            "Seed",
            "Mean_ROC_AUC",
            "SD_ROC_AUC",
            "Minimum_ROC_AUC",
            "Maximum_ROC_AUC",
            "Range_ROC_AUC",
            "CV_Percent"
        ]
    ]
    .sort_values("Seed")
    .reset_index(drop=True)
)

print(
    cv_seed_summary_table.to_string(
        index=False,
        formatters={
            "Mean_ROC_AUC": "{:.4f}".format,
            "SD_ROC_AUC": "{:.4f}".format,
            "Minimum_ROC_AUC": "{:.4f}".format,
            "Maximum_ROC_AUC": "{:.4f}".format,
            "Range_ROC_AUC": "{:.4f}".format,
            "CV_Percent": "{:.2f}".format
        }
    )
)

# ----------------------------------------------------------------------
# OVERALL 15-FOLD CV SUMMARY
# ----------------------------------------------------------------------

all_cv_roc_auc = (
    cv_fold_results_df[
        "ROC_AUC"
    ]
    .astype(float)
)

overall_cv_mean = (
    all_cv_roc_auc.mean()
)

overall_cv_sd = (
    all_cv_roc_auc.std(ddof=1)
)

overall_cv_min = (
    all_cv_roc_auc.min()
)

overall_cv_max = (
    all_cv_roc_auc.max()
)

overall_cv_range = (
    overall_cv_max -
    overall_cv_min
)

overall_cv_cv_percent = (
    overall_cv_sd /
    overall_cv_mean
) * 100

assert np.isfinite([
    overall_cv_mean,
    overall_cv_sd,
    overall_cv_min,
    overall_cv_max,
    overall_cv_range,
    overall_cv_cv_percent
]).all()

print()
print("=" * 70)
print("OVERALL 15-FOLD CV ROBUSTNESS")
print("=" * 70)

print(
    f"Overall ROC-AUC mean : "
    f"{overall_cv_mean:.4f}"
)

print(
    f"Overall ROC-AUC SD   : "
    f"{overall_cv_sd:.4f}"
)

print(
    f"Minimum ROC-AUC      : "
    f"{overall_cv_min:.4f}"
)

print(
    f"Maximum ROC-AUC      : "
    f"{overall_cv_max:.4f}"
)

print(
    f"Range                : "
    f"{overall_cv_range:.4f}"
)

print(
    f"CV (%)               : "
    f"{overall_cv_cv_percent:.2f}"
)

# ----------------------------------------------------------------------
# PRIMARY-SEED TEST RESULT
# ----------------------------------------------------------------------

primary_seed_result = (
    seed_test_results_df.loc[
        seed_test_results_df["Seed"] == PRIMARY_SEED
    ]
    .iloc[0]
)

print()
print("=" * 70)
print("PRIMARY SEED — INDEPENDENT TEST PERFORMANCE")
print("=" * 70)

for metric in PERFORMANCE_METRICS:

    print(
        f"{metric:<15}: "
        f"{primary_seed_result[metric]:.4f}"
    )

# ----------------------------------------------------------------------
# FINAL INTERPRETIVE VALIDATION
# ----------------------------------------------------------------------

print()
print("=" * 70)
print("CELL 5 — PERFORMANCE SUMMARY VALIDATION")
print("=" * 70)

print("✓ Three independent seed results summarized")
print("✓ Mean and sample SD calculated for all seven metrics")
print("✓ Minimum and maximum performance calculated")
print("✓ Performance ranges calculated")
print("✓ Coefficients of variation calculated")
print("✓ Seeds associated with maximum and minimum observed metric values identified")
print("✓ Five-fold CV summary preserved for each seed")
print("✓ All 15 CV ROC-AUC observations represented")
print("✓ Overall 15-fold CV mean and SD calculated")
print("✓ Primary seed 42 independently identified")
print("✓ Independent-test results kept separate from CV results")
print("✓ No model retrained")
print("✓ No data modified")
print("✓ Notebook 11 remains locked")
print("✓ Notebook 12 remains locked")

print()
print("=" * 70)
print("✓ CELL 5 — COMPLETED")
print("=" * 70)
```

```python
# ======================================================================
# CELL 10 — NOTEBOOK 13
# PERFORMANCE IMPROVEMENT ANALYSIS
# ======================================================================

print("=" * 70)
print("NOTEBOOK 13 — PERFORMANCE IMPROVEMENT ANALYSIS")
print("=" * 70)

import numpy as np
import pandas as pd

# ----------------------------------------------------------------------
# REQUIRED METRICS
# ----------------------------------------------------------------------

FINAL_METRICS = [
    "ROC-AUC",
    "PR-AUC",
    "F1",
    "Precision",
    "Recall",
    "Specificity",
    "Accuracy"
]

# ----------------------------------------------------------------------
# EXTRACT CORE MODEL ROWS
# ----------------------------------------------------------------------

baseline_row = (
    final_six_model_comparison_df.loc[
        final_six_model_comparison_df["Model"]
        == "Baseline XGBoost"
    ]
    .iloc[0]
)

s_xgb_row = (
    final_six_model_comparison_df.loc[
        final_six_model_comparison_df["Model"]
        == "S-XGBoost"
    ]
    .iloc[0]
)

sp_xgb_row = (
    final_six_model_comparison_df.loc[
        final_six_model_comparison_df["Model"]
        == "SP-XGBoost"
    ]
    .iloc[0]
)

# ----------------------------------------------------------------------
# CORE IMPROVEMENTS
# ----------------------------------------------------------------------

improvement_rows = []

for metric in FINAL_METRICS:

    baseline_value = float(
        baseline_row[metric]
    )

    s_value = float(
        s_xgb_row[metric]
    )

    sp_value = float(
        sp_xgb_row[metric]
    )

    improvement_rows.append({
        "Metric": metric,
        "Baseline": baseline_value,
        "S_XGBoost": s_value,
        "SP_XGBoost": sp_value,
        "S_minus_Baseline": (
            s_value - baseline_value
        ),
        "SP_minus_S": (
            sp_value - s_value
        ),
        "SP_minus_Baseline": (
            sp_value - baseline_value
        )
    })

performance_improvement_df = pd.DataFrame(
    improvement_rows
)

# ----------------------------------------------------------------------
# RELATIVE PERCENT IMPROVEMENT
# ----------------------------------------------------------------------

performance_improvement_df[
    "SP_vs_Baseline_Percent"
] = (
    performance_improvement_df[
        "SP_minus_Baseline"
    ]
    /
    performance_improvement_df[
        "Baseline"
    ]
) * 100

# ----------------------------------------------------------------------
# VALIDATION
# ----------------------------------------------------------------------

assert len(performance_improvement_df) == 7

assert np.isfinite(
    performance_improvement_df[
        [
            "Baseline",
            "S_XGBoost",
            "SP_XGBoost",
            "S_minus_Baseline",
            "SP_minus_S",
            "SP_minus_Baseline",
            "SP_vs_Baseline_Percent"
        ]
    ].to_numpy()
).all()

# ----------------------------------------------------------------------
# DISPLAY CORE IMPROVEMENTS
# ----------------------------------------------------------------------

print()
print("=" * 70)
print("BASELINE → S-XGBOOST → SP-XGBOOST")
print("=" * 70)

print(
    performance_improvement_df.to_string(
        index=False,
        formatters={
            "Baseline": "{:.4f}".format,
            "S_XGBoost": "{:.4f}".format,
            "SP_XGBoost": "{:.4f}".format,
            "S_minus_Baseline": "{:+.4f}".format,
            "SP_minus_S": "{:+.4f}".format,
            "SP_minus_Baseline": "{:+.4f}".format,
            "SP_vs_Baseline_Percent": "{:+.2f}".format
        }
    )
)

# ======================================================================
# STRONGEST CONVENTIONAL COMPARATOR
# ======================================================================

conventional_models = [
    "Logistic Regression",
    "Random Forest",
    "Support Vector Machine"
]

conventional_df = (
    final_six_model_comparison_df[
        final_six_model_comparison_df["Model"]
        .isin(conventional_models)
    ]
    .copy()
)

sp_vs_best_comparator_rows = []

for metric in FINAL_METRICS:

    best_idx = (
        conventional_df[metric]
        .astype(float)
        .idxmax()
    )

    best_model = (
        conventional_df.loc[
            best_idx,
            "Model"
        ]
    )

    best_value = float(
        conventional_df.loc[
            best_idx,
            metric
        ]
    )

    sp_value = float(
        sp_xgb_row[metric]
    )

    difference = (
        sp_value - best_value
    )

    relative_difference = (
        (difference / best_value) * 100
        if best_value != 0
        else np.nan
    )

    sp_vs_best_comparator_rows.append({
        "Metric": metric,
        "Best_Comparator": best_model,
        "Comparator_Value": best_value,
        "SP_XGBoost_Value": sp_value,
        "Absolute_Difference": difference,
        "Relative_Difference_Percent": relative_difference
    })

sp_vs_best_comparator_df = pd.DataFrame(
    sp_vs_best_comparator_rows
)

# ----------------------------------------------------------------------
# DISPLAY
# ----------------------------------------------------------------------

print()
print("=" * 70)
print("SP-XGBOOST VS STRONGEST CONVENTIONAL COMPARATOR")
print("=" * 70)

print(
    sp_vs_best_comparator_df.to_string(
        index=False,
        formatters={
            "Comparator_Value": "{:.4f}".format,
            "SP_XGBoost_Value": "{:.4f}".format,
            "Absolute_Difference": "{:+.4f}".format,
            "Relative_Difference_Percent": "{:+.2f}".format
        }
    )
)

# ======================================================================
# SAVE ARTIFACTS
# ======================================================================

PERFORMANCE_IMPROVEMENT_PATH = (
    RESULTS_DIR /
    "final_sp_xgboost_performance_improvements.csv"
)

SP_VS_COMPARATOR_PATH = (
    RESULTS_DIR /
    "sp_xgboost_vs_best_conventional_comparator.csv"
)

performance_improvement_df.to_csv(
    PERFORMANCE_IMPROVEMENT_PATH,
    index=False
)

sp_vs_best_comparator_df.to_csv(
    SP_VS_COMPARATOR_PATH,
    index=False
)

assert PERFORMANCE_IMPROVEMENT_PATH.exists()
assert SP_VS_COMPARATOR_PATH.exists()

# ----------------------------------------------------------------------
# RELOAD VALIDATION
# ----------------------------------------------------------------------

reloaded_improvements = pd.read_csv(
    PERFORMANCE_IMPROVEMENT_PATH
)

reloaded_comparator = pd.read_csv(
    SP_VS_COMPARATOR_PATH
)

assert len(reloaded_improvements) == 7
assert len(reloaded_comparator) == 7

# ======================================================================
# FINAL VALIDATION
# ======================================================================

print()
print("=" * 70)
print("CELL 10 — PERFORMANCE IMPROVEMENT VALIDATION")
print("=" * 70)

print("✓ Baseline → S-XGBoost improvements calculated")
print("✓ S-XGBoost → SP-XGBoost improvements calculated")
print("✓ Baseline → SP-XGBoost improvements calculated")
print("✓ Relative percentage improvements calculated")
print("✓ Strongest conventional comparator identified by metric")
print("✓ SP-XGBoost vs comparator differences calculated")
print("✓ Seven evaluation metrics preserved")
print("✓ Improvement tables saved")
print(f"✓ {PERFORMANCE_IMPROVEMENT_PATH}")
print(f"✓ {SP_VS_COMPARATOR_PATH}")
print("✓ Saved files successfully reloaded")
print("✓ No model retrained")
print("✓ No dataset modified")
print("✓ Notebook 11 remains locked")
print("✓ Notebook 12 remains locked")

print()
print("=" * 70)
print("✓ CELL 10 — COMPLETED")
print("=" * 70)
```
